# Using the TriNumberVersion Module in baseobjects

## Introduction

The `trinumberversion` module provides an implementation of a three-number versioning system through the `TriNumberVersion` class. This class implements a version system defined by three numbers: major, minor, and patch. The class follows the convention that the major number is more significant than the minor number, which is more significant than the patch number. This follows similar patterns like Semantic Versioning (https://semver.org/).

This tutorial will guide you through:
- Understanding the purpose and functionality of the `trinumberversion` module
- Creating and manipulating `TriNumberVersion` objects
- Comparing versions
- Converting versions to different formats
- Practical examples of using `TriNumberVersion` in your code

**Prerequisites:**
- Basic understanding of version numbering systems
- Familiarity with Python's object-oriented programming
- Knowledge of the `Version` abstract base class (see the `version_tutorial.ipynb`)

## Importing the Module

````
from baseobjects.versioning import TriNumberVersion
````

## Core Functionality

The `TriNumberVersion` class implements the `Version` abstract base class and provides a concrete implementation of a three-number versioning system. Let's explore its basic functionality.

### Creating TriNumberVersion Objects

You can create a `TriNumberVersion` object in several ways:

````
# From a string
version1 = TriNumberVersion("1.2.3")
print(f"From string: {version1}")

# From individual components
version2 = TriNumberVersion(major=2, minor=0, patch=1)
print(f"From components: {version2}")

# From a tuple or list
version3 = TriNumberVersion([3, 1, 4])
print(f"From list: {version3}")

# Default version (0.0.0)
version4 = TriNumberVersion()
print(f"Default version: {version4}")
````

### Accessing Version Components

You can access the individual components of a `TriNumberVersion` object:

````
version = TriNumberVersion("1.2.3")
print(f"Major: {version.major}")
print(f"Minor: {version.minor}")
print(f"Patch: {version.patch}")
````

### Version Comparison

One of the key features of the `TriNumberVersion` class is its support for version comparison. This allows you to compare different versions to determine which is newer or older.

````
# Create some version instances
v1 = TriNumberVersion("1.0.0")
v2 = TriNumberVersion("1.1.0")
v3 = TriNumberVersion("1.0.1")
v4 = TriNumberVersion("2.0.0")
v5 = TriNumberVersion("1.0.0")

# Compare versions
print(f"v1 == v5: {v1 == v5}")  # True (same version)
print(f"v1 != v2: {v1 != v2}")  # True (different versions)
print(f"v1 < v2: {v1 < v2}")  # True (1.0.0 < 1.1.0)
print(f"v1 < v3: {v1 < v3}")  # True (1.0.0 < 1.0.1)
print(f"v2 > v3: {v2 > v3}")  # True (1.1.0 > 1.0.1)
print(f"v4 > v2: {v4 > v2}")  # True (2.0.0 > 1.1.0)
print(f"v1 <= v5: {v1 <= v5}")  # True (same version)
print(f"v1 >= v5: {v1 >= v5}")  # True (same version)
````

### Type Conversion

The `TriNumberVersion` class provides methods for converting versions to different formats:

````
version = TriNumberVersion("1.2.3")

# Convert to different formats
print(f"String representation: {version.str()}")
print(f"List representation: {version.list()}")
print(f"Tuple representation: {version.tuple()}")

# The __str__ method also returns the string representation
print(f"String via __str__: {version!s}")
````

## Module Interaction

The `TriNumberVersion` class interacts with other modules in the baseobjects package, particularly the `Version` abstract base class from which it inherits. It also uses the `singlekwargdispatch` decorator from the `functions` module to handle different input types for the `set_version` method.

Let's see how this interaction works:

````
from baseobjects.versioning import Version
from baseobjects.functions import singlekwargdispatch

# Check if TriNumberVersion is a subclass of Version
print(f"TriNumberVersion is a subclass of Version: {issubclass(TriNumberVersion, Version)}")

# Demonstrate how singlekwargdispatch works in TriNumberVersion
# The set_version method uses singlekwargdispatch to handle different input types
version = TriNumberVersion()

# Set version from string
version.set_version("2.3.4")
print(f"After set_version with string: {version}")

# Set version from list
version.set_version([3, 4, 5])
print(f"After set_version with list: {version}")

# Set version from int (major) with optional minor and patch
version.set_version(4, minor=5, patch=6)
print(f"After set_version with int and kwargs: {version}")

# Set version with individual components
version.set_version(None, minor=7, patch=8, major=6)
print(f"After set_version with None and kwargs: {version}")
````

## Advanced Features

### Version Casting

The `TriNumberVersion` class inherits the `cast` class method from the `Version` class, which allows you to convert other objects to `TriNumberVersion` objects:

````
# Cast a string to a TriNumberVersion
version1 = TriNumberVersion.cast("1.2.3")
print(f"Cast from string: {version1}")

# Cast a list to a TriNumberVersion
version2 = TriNumberVersion.cast([2, 3, 4])
print(f"Cast from list: {version2}")

# Cast with pass_ parameter
try:
    # This will raise a TypeError because "invalid" can't be converted to a TriNumberVersion
    TriNumberVersion.cast("invalid")
except Exception as e:
    print(f"Error: {e}")

# With pass_=True, it returns the original object instead of raising an error
original = TriNumberVersion.cast("invalid", pass_=True)
print(f"Original with pass_=True: {original}")
````

### Comparing with Other Objects

The `TriNumberVersion` class can compare with objects that have a `VERSION` attribute or can be cast to a `TriNumberVersion`:

````
# Create a class with a VERSION attribute
class MyClass:
    def __init__(self, version) -> None:
        self.VERSION = TriNumberVersion(version)


# Create instances
obj1 = MyClass("1.0.0")
obj2 = MyClass("2.0.0")
version = TriNumberVersion("1.5.0")

# Compare TriNumberVersion with objects that have a VERSION attribute
print(f"version > obj1.VERSION: {version > obj1.VERSION}")  # True (1.5.0 > 1.0.0)
print(f"version < obj2.VERSION: {version < obj2.VERSION}")  # True (1.5.0 < 2.0.0)

# Compare TriNumberVersion directly with objects that have a VERSION attribute
print(f"version > obj1: {version > obj1}")  # True (1.5.0 > 1.0.0)
print(f"version < obj2: {version < obj2}")  # True (1.5.0 < 2.0.0)

# Compare with strings (cast to TriNumberVersion)
print(f"version > '1.0.0': {version > '1.0.0'}")  # True (1.5.0 > 1.0.0)
print(f"version < '2.0.0': {version < '2.0.0'}")  # True (1.5.0 < 2.0.0)
````

## Examples

### Example 1: Tracking Software Versions

One common use case for `TriNumberVersion` is tracking software versions. Let's create a simple class to represent a software product with version tracking:

````
class SoftwareProduct:
    """A simple class representing a software product with version tracking."""

    def __init__(self, name, version="0.1.0") -> None:
        self.name = name
        self.version = TriNumberVersion(version)
        self.version_history = [self.version.tuple()]

    def update_version(self, new_version) -> None:
        """Updates the product version."""
        new_ver = TriNumberVersion(new_version)
        if new_ver <= self.version:
            msg = f"New version {new_ver} must be greater than current version {self.version}"
            raise ValueError(msg)

        self.version = new_ver
        self.version_history.append(self.version.tuple())

    def is_compatible_with(self, other_product):
        """Checks if this product is compatible with another product.

        For this example, we'll say products are compatible if they have the same major version.
        """
        return self.version.major == other_product.version.major

    def __str__(self) -> str:
        return f"{self.name} v{self.version}"


# Create a software product
product = SoftwareProduct("MyApp", "1.0.0")
print(product)

# Update the version
product.update_version("1.1.0")
print(f"Updated to: {product}")

product.update_version("1.1.1")
print(f"Updated to: {product}")

product.update_version("2.0.0")
print(f"Updated to: {product}")

# View version history
print(f"Version history: {product.version_history}")

# Try to update to an older version (should raise an error)
try:
    product.update_version("1.0.0")
except ValueError as e:
    print(f"Error: {e}")

# Check compatibility
product1 = SoftwareProduct("App1", "2.1.0")
product2 = SoftwareProduct("App2", "2.3.0")
product3 = SoftwareProduct("App3", "3.0.0")

print(f"{product1.name} is compatible with {product2.name}: {product1.is_compatible_with(product2)}")
print(f"{product1.name} is compatible with {product3.name}: {product1.is_compatible_with(product3)}")
````

### Example 2: Semantic Versioning

The `TriNumberVersion` class follows a pattern similar to Semantic Versioning (SemVer). Let's create a subclass that explicitly implements the SemVer specification:

````
class SemVer(TriNumberVersion):
    """A class implementing Semantic Versioning 2.0.0 (https://semver.org/)."""

    def __init__(
        self, version=None, minor=None, patch=None, major=None, prerelease=None, build=None, init=True
    ) -> None:
        self.prerelease = prerelease
        self.build = build
        super().__init__(version=version, minor=minor, patch=patch, major=major, init=init)

    def construct(
        self, version=None, minor=None, patch=None, major=None, prerelease=None, build=None, **kwargs
    ) -> None:
        """Construct the version object based on inputs."""
        if prerelease is not None:
            self.prerelease = prerelease
        if build is not None:
            self.build = build

        if version is not None and isinstance(version, str):
            # Parse SemVer string
            parts = version.split("+")
            version_and_prerelease = parts[0]
            if len(parts) > 1:
                self.build = parts[1]

            parts = version_and_prerelease.split("-")
            version_core = parts[0]
            if len(parts) > 1:
                self.prerelease = parts[1]

            super().construct(version=version_core, minor=minor, patch=patch, major=major, **kwargs)
        else:
            super().construct(version=version, minor=minor, patch=patch, major=major, **kwargs)

    def str(self):
        """Return the string representation of the version."""
        result = f"{self.major}.{self.minor}.{self.patch}"
        if hasattr(self, "prerelease") and self.prerelease:
            result += f"-{self.prerelease}"
        if hasattr(self, "build") and self.build:
            result += f"+{self.build}"
        return result

    def __eq__(self, other):
        """Equality comparison following SemVer rules."""
        if isinstance(other, SemVer):
            # Core version must be equal
            if self.tuple() != other.tuple():
                return False

            # Pre-release versions are lower than normal versions
            if (
                hasattr(self, "prerelease")
                and self.prerelease
                and not (hasattr(other, "prerelease") and other.prerelease)
            ):
                return False
            if (
                hasattr(other, "prerelease")
                and other.prerelease
                and not (hasattr(self, "prerelease") and self.prerelease)
            ):
                return False

            # Pre-release versions are compared lexically
            if hasattr(self, "prerelease") and self.prerelease and hasattr(other, "prerelease") and other.prerelease:
                if self.prerelease != other.prerelease:
                    return False

            # Build metadata does not affect equality
            return True

        try:
            return self == self.cast(other)
        except TypeError:
            return super().__eq__(other)


# Create SemVer instances
semver1 = SemVer("1.0.0")
semver2 = SemVer("1.0.0-alpha")
semver3 = SemVer("1.0.0-beta")
semver4 = SemVer("1.0.0+build.123")
semver5 = SemVer("1.0.0-alpha+build.123")

print(f"SemVer: {semver1}")
print(f"SemVer with prerelease: {semver2}")
print(f"SemVer with different prerelease: {semver3}")
print(f"SemVer with build metadata: {semver4}")
print(f"SemVer with prerelease and build: {semver5}")

# Compare SemVer instances
print(f"semver1 == semver4: {semver1 == semver4}")  # True (build metadata doesn't affect equality)
print(f"semver1 == semver2: {semver1 == semver2}")  # False (prerelease affects equality)
print(f"semver2 == semver3: {semver2 == semver3}")  # False (different prereleases)
````

### Example 3: Version Range Checking

Another common use case is checking if a version falls within a specified range. Let's implement a simple version range checker:

````
class VersionRange:
    """A class representing a version range."""

    def __init__(self, min_version=None, max_version=None, include_min=True, include_max=False) -> None:
        """Initialize a version range.

        Args:
            min_version: The minimum version (inclusive by default)
            max_version: The maximum version (exclusive by default)
            include_min: Whether to include the minimum version
            include_max: Whether to include the maximum version
        """
        self.min_version = TriNumberVersion(min_version) if min_version is not None else None
        self.max_version = TriNumberVersion(max_version) if max_version is not None else None
        self.include_min = include_min
        self.include_max = include_max

    def contains(self, version) -> bool:
        """Check if a version is within this range."""
        if not isinstance(version, TriNumberVersion):
            version = TriNumberVersion(version)

        if self.min_version is not None:
            if self.include_min:
                if version < self.min_version:
                    return False
            else:
                if version <= self.min_version:
                    return False

        if self.max_version is not None:
            if self.include_max:
                if version > self.max_version:
                    return False
            else:
                if version >= self.max_version:
                    return False

        return True

    def __str__(self) -> str:
        """Return a string representation of the range."""
        min_bracket = "[" if self.include_min else "("
        max_bracket = "]" if self.include_max else ")"
        min_str = str(self.min_version) if self.min_version is not None else "-∞"
        max_str = str(self.max_version) if self.max_version is not None else "∞"
        return f"{min_bracket}{min_str}, {max_str}{max_bracket}"


# Create version ranges
range1 = VersionRange("1.0.0", "2.0.0")  # [1.0.0, 2.0.0)
range2 = VersionRange("1.0.0", "2.0.0", include_min=False, include_max=True)  # (1.0.0, 2.0.0]
range3 = VersionRange(max_version="3.0.0")  # (-∞, 3.0.0)
range4 = VersionRange(min_version="1.0.0")  # [1.0.0, ∞)

print(f"Range 1: {range1}")
print(f"Range 2: {range2}")
print(f"Range 3: {range3}")
print(f"Range 4: {range4}")

# Check if versions are within ranges
versions = [
    TriNumberVersion("0.9.0"),
    TriNumberVersion("1.0.0"),
    TriNumberVersion("1.5.0"),
    TriNumberVersion("2.0.0"),
    TriNumberVersion("3.0.0"),
]

for version in versions:
    print(f"Version {version}:")
    print(f"  In range1 {range1}: {range1.contains(version)}")
    print(f"  In range2 {range2}: {range2.contains(version)}")
    print(f"  In range3 {range3}: {range3.contains(version)}")
    print(f"  In range4 {range4}: {range4.contains(version)}")
````

## Troubleshooting / FAQs

### Q: How do I create a TriNumberVersion with specific major, minor, and patch values?

A: You can create a `TriNumberVersion` with specific values in several ways:

```python
# Using keyword arguments
version1 = TriNumberVersion(major=1, minor=2, patch=3)

# Using a string
version2 = TriNumberVersion("1.2.3")

# Using a list or tuple
version3 = TriNumberVersion([1, 2, 3])
```

### Q: Can I compare a TriNumberVersion with a string?

A: Yes, the `TriNumberVersion` class can compare with strings that represent valid version numbers:

```python
version = TriNumberVersion("1.2.3")
result = version > "1.0.0"  # True
```

### Q: How does version comparison work in TriNumberVersion?

A: Version comparison follows these rules:
1. Major version is compared first
2. If major versions are equal, minor versions are compared
3. If minor versions are also equal, patch versions are compared

For example:
- `1.0.0` < `2.0.0` (major version comparison)
- `1.0.0` < `1.1.0` (minor version comparison)
- `1.1.0` < `1.1.1` (patch version comparison)

### Q: What happens if I try to create a TriNumberVersion from an invalid string?

A: If you provide an invalid string that can't be parsed as a three-part version, a `TypeError` will be raised. Make sure your string is in the format "major.minor.patch" with each part being a valid integer.
