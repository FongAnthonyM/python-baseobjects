## Core Functionality

The `FunctionRegistry` class is designed to store and manage callable functions in a dictionary-like structure. It provides a convenient way to organize functions by name and access them when needed. Let's explore the core functionality and understand how it works.

### Basic Concept

At its core, `FunctionRegistry` is a specialized dictionary that only stores callable objects (functions, methods, or any callable). This makes it useful for:

1. Creating a centralized repository of functions
2. Dynamically accessing functions by name
3. Organizing related functions in a single container
4. Extracting callable attributes from objects

Let's start with a simple example of creating and using a `FunctionRegistry`:

````python
# Create a simple function registry
registry = FunctionRegistry()

# Define some functions to add to the registry
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

# Add functions to the registry using dictionary-like syntax
registry['add'] = add
registry['subtract'] = subtract
registry['multiply'] = multiply

# Access and use functions from the registry
print(f"2 + 3 = {registry['add'](2, 3)}")
print(f"5 - 2 = {registry['subtract'](5, 2)}")
print(f"4 * 6 = {registry['multiply'](4, 6)}")

# Check what functions are in the registry
print("\nFunctions in registry:", list(registry.keys()))
````

In the example above, we created a simple function registry and added three functions to it. We then accessed and used these functions by their names. The `FunctionRegistry` behaves like a dictionary, allowing us to add, access, and manage functions using familiar dictionary operations.

### Adding Functions from Different Sources

One of the powerful features of `FunctionRegistry` is its ability to extract and store callable attributes from objects. This is done through the `update_from_object` and `update_from_objects` methods.

Let's see how we can add functions from different sources:

````python
# Create a new function registry
registry = FunctionRegistry()

# 1. Adding functions directly
def greet(name):
    return f"Hello, {name}!"

registry['greet'] = greet

# 2. Adding functions from a dictionary
math_functions = {
    'square': lambda x: x**2,
    'cube': lambda x: x**3,
    'double': lambda x: x*2
}

registry.update(math_functions)

# 3. Adding methods from a class instance
class StringOperations:
    def reverse(self, text):
        return text[::-1]
    
    def capitalize_words(self, text):
        return ' '.join(word.capitalize() for word in text.split())
    
    def count_vowels(self, text):
        return sum(1 for char in text.lower() if char in 'aeiou')

string_ops = StringOperations()

# Add methods from the StringOperations instance
registry.update_from_object(string_ops)

# 4. Adding functions from multiple objects
class MathUtils:
    def is_even(self, num):
        return num % 2 == 0
    
    def is_prime(self, num):
        if num < 2:
            return False
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                return False
        return True

class TextUtils:
    def word_count(self, text):
        return len(text.split())
    
    def char_count(self, text):
        return len(text)

math_utils = MathUtils()
text_utils = TextUtils()

# Add methods from multiple objects
registry.update_from_objects(math_utils, text_utils)

# Check all functions in the registry
print("All functions in registry:", list(registry.keys()))

# Use some of the functions
print(f"\nGreeting: {registry['greet']('World')}")
print(f"Square of 5: {registry['square'](5)}")
print(f"Reverse of 'Python': {registry['reverse'](object(), 'Python')}")
print(f"Is 7 prime? {registry['is_prime'](math_utils, 7)}")
print(f"Word count in 'Hello World': {registry['word_count'](text_utils, 'Hello World')}")
````

In this example, we demonstrated four different ways to add functions to a `FunctionRegistry`:

1. Adding functions directly using dictionary-like syntax
2. Adding functions from a dictionary using the `update` method
3. Adding methods from a class instance using the `update_from_object` method
4. Adding methods from multiple objects using the `update_from_objects` method

The `update_from_object` method extracts all callable attributes from an object and adds them to the registry. The `update_from_objects` method does the same for multiple objects.

### Creating a Registry with Initial Functions

You can also create a `FunctionRegistry` with initial functions by passing them to the constructor:

````python
# Create a registry with initial functions
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

class Calculator:
    def multiply(self, a, b):
        return a * b
    
    def divide(self, a, b):
        if b == 0:
            raise ValueError("Cannot divide by zero")
        return a / b

calc = Calculator()

# Create a registry with functions from different sources
registry = FunctionRegistry(
    functions={'add': add, 'subtract': subtract},  # Dictionary of functions
    object_=calc,                                  # Object with methods
)

# Check what functions are in the registry
print("Functions in registry:", list(registry.keys()))

# Use the functions
print(f"\n10 + 5 = {registry['add'](10, 5)}")
print(f"10 - 5 = {registry['subtract'](10, 5)}")
print(f"10 * 5 = {registry['multiply'](calc, 10, 5)}")
print(f"10 / 5 = {registry['divide'](calc, 10, 5)}")
````

## Module Interaction

The `FunctionRegistry` class extends `BaseDict` from the `baseobjects.bases` module, inheriting its dictionary-like functionality. It also uses the `AnyCallable` type from `baseobjects.typing` for type hints.

Let's see how `FunctionRegistry` interacts with other components of the baseobjects package:

````python
# Create a custom registry that extends FunctionRegistry
class TypedFunctionRegistry(FunctionRegistry):
    """A function registry that only accepts functions with specific return types."""
    
    def __init__(self, return_type=None, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.return_type = return_type
    
    def __setitem__(self, key, value):
        # Check if the value is callable
        if not callable(value):
            raise TypeError(f"Value for key '{key}' is not callable")
        
        # If a return type is specified, check the function's return annotation
        if self.return_type is not None and hasattr(value, "__annotations__"):
            return_annotation = value.__annotations__.get("return")
            if return_annotation is not None and return_annotation != self.return_type:
                raise TypeError(f"Function '{key}' does not return {self.return_type.__name__}")
        
        # Add the function to the registry
        super().__setitem__(key, value)

# Create a registry that only accepts functions returning strings
string_registry = TypedFunctionRegistry(return_type=str)

# Add functions with proper return type annotations
def greet(name) -> str:
    return f"Hello, {name}!"

def format_number(num) -> str:
    return f"The number is {num}"

string_registry['greet'] = greet
string_registry['format_number'] = format_number

# Try to add a function with a different return type
try:
    def calculate_sum(a, b) -> int:
        return a + b
    
    string_registry['calculate_sum'] = calculate_sum
except TypeError as e:
    print(f"Error: {e}")

# Check what functions are in the registry
print("\nFunctions in string registry:", list(string_registry.keys()))

# Use the functions
print(f"\nGreeting: {string_registry['greet']('World')}")
print(f"Formatted number: {string_registry['format_number'](42)}")
````

In this example, we created a custom registry that extends `FunctionRegistry` to only accept functions with specific return types. This demonstrates how `FunctionRegistry` can be extended and integrated with other components of the baseobjects package.

## Advanced Features

### Using FunctionRegistry for Plugin Systems

One advanced use case for `FunctionRegistry` is implementing a simple plugin system. By registering functions from different modules or sources, you can create a flexible and extensible architecture:

````python
# Create a registry for our plugin system
plugin_registry = FunctionRegistry()

# Define some plugin functions
def text_plugin(text):
    """A plugin that processes text."""
    return text.upper()

def number_plugin(number):
    """A plugin that processes numbers."""
    return number * 2

def list_plugin(items):
    """A plugin that processes lists."""
    return sorted(items)

# Register the plugins
plugin_registry.update({
    'text_processor': text_plugin,
    'number_processor': number_plugin,
    'list_processor': list_plugin
})

# Function to process data using the appropriate plugin
def process_data(data):
    """Process data using the appropriate plugin based on data type."""
    if isinstance(data, str):
        return plugin_registry['text_processor'](data)
    elif isinstance(data, (int, float)):
        return plugin_registry['number_processor'](data)
    elif isinstance(data, list):
        return plugin_registry['list_processor'](data)
    else:
        raise TypeError(f"No plugin available for data type: {type(data).__name__}")

# Process different types of data
text_data = "hello world"
number_data = 42
list_data = [3, 1, 4, 1, 5, 9]

print(f"Processed text: {process_data(text_data)}")
print(f"Processed number: {process_data(number_data)}")
print(f"Processed list: {process_data(list_data)}")

# Add a new plugin dynamically
def dict_plugin(data):
    """A plugin that processes dictionaries."""
    return {k.upper(): v for k, v in data.items()}

plugin_registry['dict_processor'] = dict_plugin

# Update the process_data function to handle dictionaries
def process_data_v2(data):
    """Enhanced version that also handles dictionaries."""
    if isinstance(data, str):
        return plugin_registry['text_processor'](data)
    elif isinstance(data, (int, float)):
        return plugin_registry['number_processor'](data)
    elif isinstance(data, list):
        return plugin_registry['list_processor'](data)
    elif isinstance(data, dict):
        return plugin_registry['dict_processor'](data)
    else:
        raise TypeError(f"No plugin available for data type: {type(data).__name__}")

# Process a dictionary
dict_data = {'name': 'python', 'type': 'language'}
print(f"Processed dictionary: {process_data_v2(dict_data)}")
````

### Dynamic Function Selection

Another advanced use case is dynamic function selection based on runtime conditions:

````python
import random

# Create a registry of different sorting algorithms
sorting_registry = FunctionRegistry()

# Define different sorting algorithms
def bubble_sort(arr):
    """Simple bubble sort implementation."""
    result = arr.copy()
    n = len(result)
    for i in range(n):
        for j in range(0, n - i - 1):
            if result[j] > result[j + 1]:
                result[j], result[j + 1] = result[j + 1], result[j]
    return result

def quick_sort(arr):
    """Quick sort implementation."""
    if len(arr) <= 1:
        return arr
    result = arr.copy()
    pivot = result[len(result) // 2]
    left = [x for x in result if x < pivot]
    middle = [x for x in result if x == pivot]
    right = [x for x in result if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

def merge_sort(arr):
    """Merge sort implementation."""
    if len(arr) <= 1:
        return arr
    
    result = arr.copy()
    mid = len(result) // 2
    left = merge_sort(result[:mid])
    right = merge_sort(result[mid:])
    
    # Merge
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

# Register the sorting algorithms
sorting_registry.update({
    'bubble_sort': bubble_sort,
    'quick_sort': quick_sort,
    'merge_sort': merge_sort
})

# Function to select the appropriate sorting algorithm based on input size
def smart_sort(arr):
    """Select the appropriate sorting algorithm based on input size."""
    if len(arr) < 10:
        # For small arrays, bubble sort is simple and efficient enough
        return sorting_registry['bubble_sort'](arr)
    elif len(arr) < 1000:
        # For medium-sized arrays, quick sort is generally faster
        return sorting_registry['quick_sort'](arr)
    else:
        # For large arrays, merge sort has guaranteed O(n log n) performance
        return sorting_registry['merge_sort'](arr)

# Test with different array sizes
small_array = [3, 1, 4, 1, 5]
medium_array = [random.randint(1, 100) for _ in range(100)]
large_array = [random.randint(1, 1000) for _ in range(1000)]

print(f"Sorted small array: {smart_sort(small_array)}")
print(f"First 10 elements of sorted medium array: {smart_sort(medium_array)[:10]}")
print(f"First 10 elements of sorted large array: {smart_sort(large_array)[:10]}")
````

### Example 1: Command Pattern Implementation

The Command pattern is a behavioral design pattern that turns a request into a stand-alone object containing all information about the request. This transformation lets you pass requests as method arguments, delay or queue a request's execution, and support undoable operations.

Let's implement a simple command pattern using `FunctionRegistry`:

````python
# Create a registry for commands
command_registry = FunctionRegistry()

# Define some commands
def save_command(filename, content):
    """Command to save content to a file."""
    print(f"Saving content to {filename}...")
    # In a real application, this would actually save to a file
    return f"Saved {len(content)} characters to {filename}"

def load_command(filename):
    """Command to load content from a file."""
    print(f"Loading content from {filename}...")
    # In a real application, this would actually load from a file
    return f"Loaded content from {filename}"

def process_command(text):
    """Command to process text."""
    print(f"Processing text: {text[:20]}...")
    return text.upper()

# Register the commands
command_registry.update({
    'save': save_command,
    'load': load_command,
    'process': process_command
})

# Command executor
class CommandExecutor:
    def __init__(self, registry):
        self.registry = registry
        self.history = []
    
    def execute(self, command_name, *args, **kwargs):
        """Execute a command by name."""
        if command_name not in self.registry:
            raise ValueError(f"Unknown command: {command_name}")
        
        # Execute the command
        result = self.registry[command_name](*args, **kwargs)
        
        # Record the command in history
        self.history.append((command_name, args, kwargs, result))
        
        return result
    
    def show_history(self):
        """Show the command execution history."""
        print("\nCommand History:")
        for i, (cmd, args, kwargs, result) in enumerate(self.history, 1):
            args_str = ', '.join(str(arg) for arg in args)
            kwargs_str = ', '.join(f"{k}={v}" for k, v in kwargs.items())
            all_args = ', '.join(filter(None, [args_str, kwargs_str]))
            print(f"{i}. {cmd}({all_args}) => {result}")

# Create a command executor
executor = CommandExecutor(command_registry)

# Execute some commands
executor.execute('process', "Hello, world!")
executor.execute('save', "document.txt", "This is some content to save.")
executor.execute('load', "document.txt")

# Show command history
executor.history

# Add a new command dynamically
def delete_command(filename):
    """Command to delete a file."""
    print(f"Deleting {filename}...")
    # In a real application, this would actually delete the file
    return f"Deleted {filename}"

command_registry['delete'] = delete_command

# Execute the new command
executor.execute('delete', "old_file.txt")

# Show updated history
executor.show_history()
````

### Example 2: Strategy Pattern Implementation

The Strategy pattern is a behavioral design pattern that lets you define a family of algorithms, put each of them into a separate class, and make their objects interchangeable. Let's implement a simple strategy pattern using `FunctionRegistry`:

````python
# Create a registry for payment strategies
payment_registry = FunctionRegistry()

# Define payment strategies
def credit_card_payment(amount, card_number, expiry, cvv):
    """Process a credit card payment."""
    # In a real application, this would integrate with a payment gateway
    print(f"Processing credit card payment of ${amount:.2f}")
    print(f"Card: {card_number[:4]}...{card_number[-4:]} (Exp: {expiry})")
    return f"Credit card payment of ${amount:.2f} successful"

def paypal_payment(amount, email):
    """Process a PayPal payment."""
    # In a real application, this would integrate with PayPal API
    print(f"Processing PayPal payment of ${amount:.2f}")
    print(f"PayPal account: {email}")
    return f"PayPal payment of ${amount:.2f} successful"

def bank_transfer_payment(amount, account_number, routing_number):
    """Process a bank transfer payment."""
    # In a real application, this would integrate with banking APIs
    print(f"Processing bank transfer of ${amount:.2f}")
    print(f"Account: {account_number}, Routing: {routing_number}")
    return f"Bank transfer of ${amount:.2f} successful"

# Register the payment strategies
payment_registry.update({
    'credit_card': credit_card_payment,
    'paypal': paypal_payment,
    'bank_transfer': bank_transfer_payment
})

# Payment processor using the strategy pattern
class PaymentProcessor:
    def __init__(self, registry):
        self.registry = registry
    
    def process_payment(self, method, amount, **kwargs):
        """Process a payment using the specified method."""
        if method not in self.registry:
            raise ValueError(f"Unsupported payment method: {method}")
        
        return self.registry[method](amount, **kwargs)

# Create a payment processor
processor = PaymentProcessor(payment_registry)

# Process different types of payments
print(processor.process_payment(
    'credit_card', 
    99.99, 
    card_number="4111111111111111", 
    expiry="12/25", 
    cvv="123"
))

print("\n" + processor.process_payment(
    'paypal', 
    49.99, 
    email="customer@example.com"
))

print("\n" + processor.process_payment(
    'bank_transfer', 
    199.99, 
    account_number="12345678", 
    routing_number="087654321"
))

# Add a new payment method dynamically
def crypto_payment(amount, wallet_address, currency="BTC"):
    """Process a cryptocurrency payment."""
    print(f"Processing {currency} payment of ${amount:.2f}")
    print(f"Wallet: {wallet_address}")
    return f"{currency} payment of ${amount:.2f} successful"

payment_registry['crypto'] = crypto_payment

# Process a payment with the new method
print("\n" + processor.process_payment(
    'crypto', 
    299.99, 
    wallet_address="1A1zP1eP5QGefi2DMPTfTL5SLmv7DivfNa",
    currency="ETH"
))
````

## API Highlights

The `FunctionRegistry` class provides the following key features:

- **Dictionary-like Interface**: Inherits from `BaseDict`, providing familiar dictionary operations
- **Function Storage**: Stores callable objects (functions, methods, lambdas, etc.)
- **Object Integration**: Methods to extract callable attributes from objects
- **Initialization Options**: Can be created with initial functions from various sources

Key methods:
- `__init__(functions=None, object_=None, objects=None, ...)`: Constructor with options for initial functions
- `construct(functions=None, object_=None, objects=None, ...)`: Constructor method with the same options
- `update_from_object(object_)`: Extracts callable attributes from an object and adds them to the registry
- `update_from_objects(*args)`: Extracts callable attributes from multiple objects and adds them to the registry

For the full API documentation, refer to the baseobjects documentation.

## Troubleshooting / FAQs

### Q: Why isn't my function being added to the registry?

A: The `FunctionRegistry` only stores callable objects. Make sure the object you're trying to add is callable (a function, method, or an object with a `__call__` method). You can check if an object is callable using the `callable()` function.

### Q: How do I handle methods that require the 'self' parameter?

A: When you extract methods from an object using `update_from_object`, the methods are bound to the object, so you don't need to worry about the `self` parameter. However, if you're adding unbound methods, you'll need to provide the instance as the first argument when calling the method.

### Q: Can I store class methods or static methods in a FunctionRegistry?

A: Yes, you can store class methods and static methods in a `FunctionRegistry`. When extracted from a class using `update_from_object`, class methods will be bound to the class, and static methods will be accessible as regular functions.

### Q: How do I handle functions with the same name from different objects?

A: When adding functions from multiple objects, functions with the same name will overwrite each other. If you need to keep functions with the same name from different objects, you can use a naming convention or prefix/suffix to distinguish them:

```python
# Add methods with prefixes
for name in dir(obj1):
    attr = getattr(obj1, name)
    if callable(attr):
        registry[f"obj1_{name}"] = attr

for name in dir(obj2):
    attr = getattr(obj2, name)
    if callable(attr):
        registry[f"obj2_{name}"] = attr
```
