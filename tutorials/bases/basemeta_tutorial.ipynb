The `BaseMeta` class extends Python's `ABCMeta` (Abstract Base Class Metaclass) to add proper support for copying and deep copying of metaclass instances (i.e., classes created with this metaclass).

## Importing the Module

```
from baseobjects.bases.basemeta import BaseMeta
import copy
from abc import abstractmethod
```

## Core Functionality

The `BaseMeta` class is an abstract metaclass that inherits from Python's `ABCMeta` and adds functionality for copying and deep copying metaclass objects. Its primary purpose is to serve as a foundation for other metaclasses, ensuring that classes created with these metaclasses can be properly copied.

### Key Features

1. **Copy Protocol Implementation**: Properly implements Python's copy protocol through `__copy__` and `__deepcopy__` methods
2. **Metaclass Inheritance**: Inherits from `ABCMeta`, allowing for the creation of abstract base classes with metaclass functionality
3. **Proper Handling of Class Attributes**: Ensures that class attributes are properly copied during copying operations

Let's create a simple class that uses `BaseMeta` as its metaclass:

```
class MyClass(metaclass=BaseMeta):
    """A simple class using BaseMeta as its metaclass."""

    class_attribute = "This is a class attribute"

    def __init__(self, instance_attribute) -> None:
        self.instance_attribute = instance_attribute

    def get_attributes(self):
        return {
            "class_attribute": self.class_attribute,
            "instance_attribute": self.instance_attribute,
        }


# Create an instance of MyClass
obj = MyClass("This is an instance attribute")
print(f"Original class: {MyClass}")
print(f"Class attributes: {MyClass.class_attribute}")
print(f"Instance attributes: {obj.get_attributes()}")
```

### Copying Classes

One of the key features of `BaseMeta` is its implementation of Python's copy protocol for metaclass instances (i.e., classes). Let's see how the `copy` module works with classes created with `BaseMeta`:

```
# Create a copy of the class
MyClass_copy = copy.copy(MyClass)

print(f"Original class: {MyClass}")
print(f"Copied class: {MyClass_copy}")
print(f"Are they the same class? {MyClass is MyClass_copy}")
print(f"Do they have the same class attribute? {MyClass.class_attribute is MyClass_copy.class_attribute}")

# Modify the class attribute of the copy
MyClass_copy.class_attribute = "Modified class attribute"

print("\nAfter modification:")
print(f"Original class attribute: {MyClass.class_attribute}")
print(f"Copied class attribute: {MyClass_copy.class_attribute}")

# Create instances of both classes
obj1 = MyClass("Instance of original class")
obj2 = MyClass_copy("Instance of copied class")

print(f"\nInstance of original class: {obj1.get_attributes()}")
print(f"Instance of copied class: {obj2.get_attributes()}")
```

### Deep Copying Classes

For a completely independent copy, including all nested objects, we can use the `deepcopy` function:

```
# Create a deep copy of the class
MyClass_deepcopy = copy.deepcopy(MyClass)

print(f"Original class: {MyClass}")
print(f"Deep copied class: {MyClass_deepcopy}")
print(f"Are they the same class? {MyClass is MyClass_deepcopy}")

# Modify the class attribute of the deep copy
MyClass_deepcopy.class_attribute = "Deep copied class attribute"

print("\nAfter modification:")
print(f"Original class attribute: {MyClass.class_attribute}")
print(f"Deep copied class attribute: {MyClass_deepcopy.class_attribute}")

# Create instances of both classes
obj1 = MyClass("Instance of original class")
obj2 = MyClass_deepcopy("Instance of deep copied class")

print(f"\nInstance of original class: {obj1.get_attributes()}")
print(f"Instance of deep copied class: {obj2.get_attributes()}")
```

## Module Interaction

The `BaseMeta` module interacts with other modules in the Python standard library, particularly `abc` and `copy`. It also serves as a foundation for other metaclasses in the baseobjects package.

### Interaction with ABCMeta

`BaseMeta` inherits from `ABCMeta`, which means it can be used to create abstract base classes with metaclass functionality:

```
class AbstractBase(metaclass=BaseMeta):
    """An abstract base class using BaseMeta as its metaclass."""

    @abstractmethod
    def abstract_method(self):
        """This method must be implemented by subclasses."""


# Try to instantiate the abstract base class
try:
    obj = AbstractBase()
except TypeError as e:
    print(f"Error: {e}")


# Create a concrete subclass
class ConcreteClass(AbstractBase):
    def abstract_method(self) -> str:
        return "Implementation of abstract_method"


# Instantiate the concrete subclass
obj = ConcreteClass()
print(f"Result of abstract_method: {obj.abstract_method()}")
```

### Interaction with Copy Module

`BaseMeta` implements the copy protocol, which means it works seamlessly with Python's `copy` module:

```
# Create a class with BaseMeta as its metaclass
class ClassWithMeta(metaclass=BaseMeta):
    class_attribute = {"key": "value"}

    def __init__(self, name) -> None:
        self.name = name


# Create a copy of the class
ClassWithMeta_copy = copy.copy(ClassWithMeta)

# Modify the class attribute of the copy
ClassWithMeta_copy.class_attribute["key"] = "modified value"

print(f"Original class attribute: {ClassWithMeta.class_attribute}")
print(f"Copied class attribute: {ClassWithMeta_copy.class_attribute}")
print(f"Are they the same dictionary? {ClassWithMeta.class_attribute is ClassWithMeta_copy.class_attribute}")

# Create a deep copy of the class
ClassWithMeta_deepcopy = copy.deepcopy(ClassWithMeta)

# Modify the class attribute of the deep copy
ClassWithMeta_deepcopy.class_attribute["key"] = "deep copied value"

print("\nAfter deep copy modification:")
print(f"Original class attribute: {ClassWithMeta.class_attribute}")
print(f"Deep copied class attribute: {ClassWithMeta_deepcopy.class_attribute}")
print(f"Are they the same dictionary? {ClassWithMeta.class_attribute is ClassWithMeta_deepcopy.class_attribute}")
```

## Advanced Features

The `BaseMeta` class provides several advanced features that make it powerful for creating metaclasses.

### Creating Custom Metaclasses

You can create your own metaclasses by inheriting from `BaseMeta`:

```
class CustomMeta(BaseMeta):
    """A custom metaclass that adds a class method to its instances."""

    def get_class_name(self) -> str:
        """A class method that returns the class name."""
        return f"Class name: {self.__name__}"


# Create a class with CustomMeta as its metaclass
class ClassWithCustomMeta(metaclass=CustomMeta):
    pass


# Use the class method added by the metaclass
print(ClassWithCustomMeta.get_class_name())

# Create a copy of the class
ClassWithCustomMeta_copy = copy.copy(ClassWithCustomMeta)

# Verify that the copied class also has the class method
print(ClassWithCustomMeta_copy.get_class_name())
```

### Metaclass Inheritance

Classes that inherit from a class with a metaclass will also use that metaclass:

```
class BaseWithMeta(metaclass=BaseMeta):
    """A base class with BaseMeta as its metaclass."""
    base_attribute = "Base attribute"


class ChildClass(BaseWithMeta):
    """A child class that inherits from BaseWithMeta."""
    child_attribute = "Child attribute"


# Check the metaclass of the child class
print(f"ChildClass's metaclass: {type(ChildClass)}")
print(f"Is ChildClass's metaclass BaseMeta? {isinstance(type(ChildClass), BaseMeta)}")

# Create a copy of the child class
ChildClass_copy = copy.copy(ChildClass)

print(f"\nOriginal child class: {ChildClass}")
print(f"Copied child class: {ChildClass_copy}")
print(f"Are they the same class? {ChildClass is ChildClass_copy}")

# Modify attributes of the copied class
ChildClass_copy.base_attribute = "Modified base attribute"
ChildClass_copy.child_attribute = "Modified child attribute"

print("\nAfter modification:")
print(f"Original base attribute: {ChildClass.base_attribute}")
print(f"Copied base attribute: {ChildClass_copy.base_attribute}")
print(f"Original child attribute: {ChildClass.child_attribute}")
print(f"Copied child attribute: {ChildClass_copy.child_attribute}")
```

## Examples

Let's explore some practical examples of using the `BaseMeta` module.

### Creating a Registry Metaclass

A common use case for metaclasses is creating a registry of classes. Let's create a metaclass that registers all classes created with it:

```
class RegistryMeta(BaseMeta):
    """A metaclass that maintains a registry of all classes created with it."""

    # Class attribute to store the registry
    registry = {}

    def __new__(mcs, name, bases, namespace):
        # Create the class
        cls = super().__new__(mcs, name, bases, namespace)

        # Register the class
        mcs.registry[name] = cls

        return cls


# Create some classes with RegistryMeta
class RegisteredClass1(metaclass=RegistryMeta):
    pass


class RegisteredClass2(metaclass=RegistryMeta):
    pass


# Check the registry
print("Registry contents:")
for name, cls in RegistryMeta.registry.items():
    print(f"- {name}: {cls}")

# Create a copy of the metaclass
RegistryMeta_copy = copy.copy(RegistryMeta)

# Check if the registry was copied
print("\nCopied registry contents:")
for name, cls in RegistryMeta_copy.registry.items():
    print(f"- {name}: {cls}")


# Add a new class to the original registry
class RegisteredClass3(metaclass=RegistryMeta):
    pass


print("\nOriginal registry after adding a new class:")
for name, cls in RegistryMeta.registry.items():
    print(f"- {name}: {cls}")

print("\nCopied registry after adding a new class:")
for name, cls in RegistryMeta_copy.registry.items():
    print(f"- {name}: {cls}")
```

### Creating a Validation Metaclass

Another common use case for metaclasses is validating class definitions. Let's create a metaclass that ensures all methods have docstrings:

```
class DocstringMeta(BaseMeta):
    """A metaclass that ensures all methods have docstrings."""

    def __new__(mcs, name, bases, namespace):
        # Check all methods for docstrings
        for key, value in namespace.items():
            if callable(value) and not key.startswith("__"):
                if not value.__doc__:
                    msg = f"Method {key} in class {name} must have a docstring"
                    raise TypeError(msg)

        # Create the class
        return super().__new__(mcs, name, bases, namespace)


# Create a class with DocstringMeta
try:
    class ClassWithoutDocstrings(metaclass=DocstringMeta):
        def method_without_docstring(self) -> None:
            pass
except TypeError as e:
    print(f"Error: {e}")


# Create a class with proper docstrings
class ClassWithDocstrings(metaclass=DocstringMeta):
    def method_with_docstring(self) -> None:
        """This method has a docstring."""


# Create a copy of the class
ClassWithDocstrings_copy = copy.copy(ClassWithDocstrings)

# Check if the copied class still enforces docstrings
try:
    # Try to add a method without a docstring to the copied class
    ClassWithDocstrings_copy.method_without_docstring = lambda self: None

    # Create a new class that inherits from the copied class
    class ChildWithoutDocstrings(ClassWithDocstrings_copy):
        def another_method_without_docstring(self) -> None:
            pass
except TypeError as e:
    print(f"Error: {e}")
```

## Troubleshooting / FAQs

### Q: Why would I need to copy a class?

A: Copying classes can be useful in various scenarios, such as:
- Creating variations of a class at runtime
- Implementing class factories that generate customized classes
- Testing different configurations of a class without modifying the original
- Implementing undo/redo functionality for class modifications

### Q: What's the difference between a shallow copy and a deep copy of a class?

A: When you create a shallow copy of a class, the class attributes that are mutable objects (like dictionaries or lists) are shared between the original and the copy. When you create a deep copy, all class attributes are recursively copied, resulting in completely independent objects.

### Q: Can I use BaseMeta with other metaclasses?

A: Combining multiple metaclasses can be challenging due to the metaclass conflict issue in Python. If you need to combine `BaseMeta` with another metaclass, you might need to create a new metaclass that inherits from both and resolves any conflicts in method resolution.
