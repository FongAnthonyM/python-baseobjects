## Importing the Module

```
from baseobjects.bases.collections.basedict import BaseDict
import copy
```

## Core Functionality

The `BaseDict` class is an abstract class that inherits from both `BaseObject` and `UserDict`. It combines the dictionary-like behavior of `UserDict` with the enhanced functionality of `BaseObject`, such as proper copying and deep copying support.

### Key Features

1. **Dictionary-Like Behavior**: Provides all the standard dictionary methods and operations
2. **Enhanced Copying**: Properly implements Python's copy protocol through inheritance from `BaseObject`
3. **Flexible Initialization**: Accepts an optional dictionary-like object to initialize the contents

Let's create a simple `BaseDict` instance and explore its functionality:

```
# Create a BaseDict instance
base_dict = BaseDict({"key1": "value1", "key2": "value2", "key3": "value3"})
print(f"BaseDict: {base_dict}")

# Access values using dictionary syntax
print(f"base_dict['key1']: {base_dict['key1']}")
print(f"base_dict['key2']: {base_dict['key2']}")

# Check if a key exists
print(f"'key3' in base_dict: {'key3' in base_dict}")
print(f"'key4' in base_dict: {'key4' in base_dict}")

# Get the number of items
print(f"len(base_dict): {len(base_dict)}")
```

### Dictionary Methods

`BaseDict` inherits all the standard dictionary methods from `UserDict`:

```
# Get all keys
print(f"base_dict.keys(): {base_dict.keys()}")

# Get all values
print(f"base_dict.values(): {base_dict.values()}")

# Get all items
print(f"base_dict.items(): {base_dict.items()}")

# Get a value with a default if the key doesn't exist
print(f"base_dict.get('key3'): {base_dict.get('key3')}")
print(f"base_dict.get('key4', 'default'): {base_dict.get('key4', 'default')}")
```

### Modifying the Dictionary

`BaseDict` supports all the standard dictionary modification operations:

```
# Add a new key-value pair
base_dict["key4"] = "value4"
print(f"After adding 'key4': {base_dict}")

# Update multiple key-value pairs
base_dict.update({"key5": "value5", "key6": "value6"})
print(f"After update: {base_dict}")

# Remove a key-value pair
del base_dict["key6"]
print(f"After deleting 'key6': {base_dict}")

# Pop a key-value pair
value = base_dict.pop("key5")
print(f"Popped value: {value}")
print(f"After pop: {base_dict}")

# Clear all key-value pairs
base_dict_copy = base_dict.copy()  # Make a copy to preserve the original
base_dict_copy.clear()
print(f"After clear: {base_dict_copy}")
print(f"Original: {base_dict}")
```

## Module Interaction

The `BaseDict` module interacts with other modules in the baseobjects package, particularly `BaseObject`. These interactions provide enhanced functionality for dictionary-like objects.

### Interaction with BaseObject

`BaseDict` inherits from `BaseObject`, which means it also inherits all the functionality of `BaseObject`, including copying and deep copying:

```
# Create a copy of the BaseDict instance
base_dict_copy = base_dict.copy()
print(f"Original: {base_dict}")
print(f"Copy: {base_dict_copy}")
print(f"Are they the same object? {base_dict is base_dict_copy}")
print(f"Do they have the same data? {base_dict.data is base_dict_copy.data}")

# Modify the copy
base_dict_copy["key1"] = "modified value"
print(f"\nAfter modifying the copy:")
print(f"Original: {base_dict}")
print(f"Copy: {base_dict_copy}")
```

### Deep Copying

`BaseDict` also supports deep copying through inheritance from `BaseObject`:

```
# Create a BaseDict with nested dictionaries
nested_dict = BaseDict({
    "simple_key": "simple_value",
    "dict_key": {"nested_key1": "nested_value1", "nested_key2": "nested_value2"},
    "list_key": [1, 2, 3, 4, 5]
})
print(f"Nested dict: {nested_dict}")

# Create a deep copy
nested_dict_deepcopy = copy.deepcopy(nested_dict)
print(f"Deep copy: {nested_dict_deepcopy}")
print(f"Are they the same object? {nested_dict is nested_dict_deepcopy}")
print(f"Is the nested dict the same object? {nested_dict['dict_key'] is nested_dict_deepcopy['dict_key']}")
print(f"Is the list the same object? {nested_dict['list_key'] is nested_dict_deepcopy['list_key']}")

# Modify the nested dictionary in the deep copy
nested_dict_deepcopy["dict_key"]["nested_key1"] = "modified nested value"
nested_dict_deepcopy["list_key"].append(6)

print(f"\nAfter modifying the deep copy:")
print(f"Original nested dict: {nested_dict['dict_key']}")
print(f"Original list: {nested_dict['list_key']}")
print(f"Deep copy nested dict: {nested_dict_deepcopy['dict_key']}")
print(f"Deep copy list: {nested_dict_deepcopy['list_key']}")
```

## Advanced Features

The `BaseDict` class provides several advanced features that make it powerful for creating custom dictionary-like objects.

### Creating Custom Dictionary Classes

You can create your own dictionary classes by inheriting from `BaseDict`:

```
class CaseInsensitiveDict(BaseDict):
    """A dictionary that is case-insensitive for string keys."""
    
    def __getitem__(self, key):
        if isinstance(key, str):
            for k in self.data:
                if isinstance(k, str) and k.lower() == key.lower():
                    return self.data[k]
            raise KeyError(key)
        return self.data[key]
    
    def __setitem__(self, key, value):
        if isinstance(key, str):
            for k in list(self.data.keys()):
                if isinstance(k, str) and k.lower() == key.lower():
                    del self.data[k]
                    break
        self.data[key] = value
    
    def __contains__(self, key):
        if isinstance(key, str):
            for k in self.data:
                if isinstance(k, str) and k.lower() == key.lower():
                    return True
            return False
        return key in self.data

# Create a CaseInsensitiveDict instance
case_insensitive_dict = CaseInsensitiveDict({"Name": "John", "Age": 30, "City": "New York"})
print(f"CaseInsensitiveDict: {case_insensitive_dict}")

# Access values using different case
print(f"case_insensitive_dict['name']: {case_insensitive_dict['name']}")
print(f"case_insensitive_dict['AGE']: {case_insensitive_dict['AGE']}")
print(f"case_insensitive_dict['city']: {case_insensitive_dict['city']}")

# Check if a key exists
print(f"'NAME' in case_insensitive_dict: {'NAME' in case_insensitive_dict}")
print(f"'address' in case_insensitive_dict: {'address' in case_insensitive_dict}")

# Add a new key-value pair
case_insensitive_dict["address"] = "123 Main St"
print(f"After adding 'address': {case_insensitive_dict}")

# Update an existing key with different case
case_insensitive_dict["NAME"] = "Jane"
print(f"After updating 'NAME': {case_insensitive_dict}")
```

### Adding Custom Functionality

You can add custom functionality to your dictionary classes:

```
class DefaultValueDict(BaseDict):
    """A dictionary that returns a default value for missing keys."""
    
    def __init__(self, default_value=None, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.default_value = default_value
    
    def __getitem__(self, key):
        try:
            return super().__getitem__(key)
        except KeyError:
            return self.default_value
    
    def get(self, key, default=None):
        try:
            return super().__getitem__(key)
        except KeyError:
            return default if default is not None else self.default_value

# Create a DefaultValueDict instance
default_dict = DefaultValueDict("N/A", {"name": "John", "age": 30})
print(f"DefaultValueDict: {default_dict}")

# Access existing keys
print(f"default_dict['name']: {default_dict['name']}")
print(f"default_dict['age']: {default_dict['age']}")

# Access non-existing keys
print(f"default_dict['city']: {default_dict['city']}")
print(f"default_dict['country']: {default_dict['country']}")

# Use the get method
print(f"default_dict.get('name'): {default_dict.get('name')}")
print(f"default_dict.get('address'): {default_dict.get('address')}")
print(f"default_dict.get('phone', 'Unknown'): {default_dict.get('phone', 'Unknown')}")
```

### Creating a Configuration Dictionary

`BaseDict` is useful for creating configuration dictionaries with enhanced functionality:

```
class ConfigDict(BaseDict):
    """A dictionary for storing configuration values with type checking."""
    
    def __init__(self, config_schema=None, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.schema = config_schema or {}
    
    def __setitem__(self, key, value):
        if key in self.schema:
            expected_type = self.schema[key]
            if not isinstance(value, expected_type):
                raise TypeError(f"Value for '{key}' must be of type {expected_type.__name__}, got {type(value).__name__}")
        super().__setitem__(key, value)
    
    def validate(self):
        """Validate that all required keys are present and have the correct type."""
        for key, expected_type in self.schema.items():
            if key not in self:
                raise KeyError(f"Missing required configuration key: '{key}'")
            if not isinstance(self[key], expected_type):
                raise TypeError(f"Value for '{key}' must be of type {expected_type.__name__}, got {type(self[key]).__name__}")
        return True

# Define a configuration schema
config_schema = {
    "host": str,
    "port": int,
    "debug": bool,
    "timeout": (int, float),  # Accept either int or float
    "max_connections": int
}

# Create a ConfigDict instance
config = ConfigDict(config_schema)

# Set valid configuration values
config["host"] = "localhost"
config["port"] = 8080
config["debug"] = True
config["timeout"] = 30.5
config["max_connections"] = 100

print(f"Configuration: {config}")

# Try to set an invalid value
try:
    config["port"] = "8080"  # Should be an int
except TypeError as e:
    print(f"Error: {e}")

# Validate the configuration
try:
    config.validate()
    print("Configuration is valid")
except (KeyError, TypeError) as e:
    print(f"Validation error: {e}")
```

### Implementing a Frequency Counter

`BaseDict` can be used to implement a frequency counter:

```
class FrequencyCounter(BaseDict):
    """A dictionary that counts the frequency of items."""
    
    def add(self, item):
        """Add an item to the counter."""
        self[item] = self.get(item, 0) + 1
    
    def add_all(self, items):
        """Add multiple items to the counter."""
        for item in items:
            self.add(item)
    
    def most_common(self, n=None):
        """Return the n most common items and their counts."""
        sorted_items = sorted(self.items(), key=lambda x: x[1], reverse=True)
        return sorted_items[:n] if n is not None else sorted_items
    
    def least_common(self, n=None):
        """Return the n least common items and their counts."""
        sorted_items = sorted(self.items(), key=lambda x: x[1])
        return sorted_items[:n] if n is not None else sorted_items

# Create a FrequencyCounter instance
counter = FrequencyCounter()

# Add items to the counter
text = "the quick brown fox jumps over the lazy dog"
counter.add_all(text.split())

print(f"Frequency counter: {counter}")

# Get the most common items
most_common = counter.most_common(3)
print(f"Most common items: {most_common}")

# Get the least common items
least_common = counter.least_common(3)
print(f"Least common items: {least_common}")
```

## Troubleshooting / FAQs

### Q: When should I use BaseDict instead of a regular dictionary?

A: `BaseDict` is useful when you need a dictionary-like object with enhanced functionality, such as proper copying and deep copying support, or when you want to create a custom dictionary class with additional methods and behaviors.

### Q: How does BaseDict handle copying and deep copying?

A: `BaseDict` inherits from `BaseObject`, which properly implements Python's copy protocol through `__copy__` and `__deepcopy__` methods. This ensures that both the dictionary and its contents are properly copied.

### Q: Can I use BaseDict with non-hashable keys?

A: Like regular dictionaries, `BaseDict` requires keys to be hashable. If you need to use non-hashable objects as keys, you'll need to implement a custom hashing mechanism or use a different data structure.
