# Using the UnionRecursive Module in baseobjects

## Introduction

The `unionrecursive` module provides functionality for recursively unioning (merging) mapping objects and their contained mappings. This is useful when you need to combine nested dictionaries or other mapping types while preserving the hierarchical structure.

The module contains a single function, `union_recursive`, which takes two mapping objects and returns a new mapping that is the recursive union of the two. Unlike the standard dictionary update method, which only updates at the top level, `union_recursive` will recursively update nested mappings as well.

This tutorial will guide you through:
- Understanding the purpose and functionality of the `unionrecursive` module
- Using the `union_recursive` function with different mapping types
- Comparing recursive union with standard dictionary operations
- Practical examples and use cases

**Prerequisites:**
- Basic understanding of Python dictionaries and other mapping types
- Familiarity with dictionary operations like update and merge
- Knowledge of nested data structures

## Importing the Module

````
from baseobjects.operations.unionrecursive import union_recursive
````

## Core Functionality

The `union_recursive` function creates a new mapping that is the recursive union of two input mappings. Let's explore its basic functionality.

### Basic Usage

Let's start with a simple example to see how the function works:

````
# Create two dictionaries
dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"b": 20, "c": 30, "d": 40}

# Union them recursively
result = union_recursive(dict1, dict2)

# Display the results
print(f"dict1: {dict1}")
print(f"dict2: {dict2}")
print(f"union_recursive(dict1, dict2): {result}")
````

In this example, the function returns a new dictionary that contains all keys from both input dictionaries. For keys that exist in both dictionaries, the values from the second dictionary (`dict2`) take precedence.

Now, let's try with nested dictionaries:

````
# Create two nested dictionaries
nested_dict1 = {
    "a": 1,
    "b": {
        "x": 10,
        "y": 20,
    },
    "c": {
        "p": 100,
        "q": 200,
    },
}

nested_dict2 = {
    "b": {
        "y": 30,
        "z": 40,
    },
    "c": 300,
    "d": 400,
}

# Union them recursively
nested_result = union_recursive(nested_dict1, nested_dict2)

# Display the results
print(f"nested_dict1: {nested_dict1}")
print(f"nested_dict2: {nested_dict2}")
print(f"union_recursive(nested_dict1, nested_dict2): {nested_result}")
````

In this example, the function recursively merges the nested dictionaries. For the key `"b"`, which exists in both dictionaries and has a dictionary value in both, the function recursively merges the nested dictionaries. For the key `"c"`, which exists in both dictionaries but has a dictionary value in `nested_dict1` and a non-dictionary value in `nested_dict2`, the value from `nested_dict2` takes precedence.

### Comparison with Standard Dictionary Operations

Let's compare `union_recursive` with standard dictionary operations:

````
# Create two nested dictionaries
dict1 = {
    "a": 1,
    "b": {
        "x": 10,
        "y": 20,
    },
}

dict2 = {
    "b": {
        "y": 30,
        "z": 40,
    },
    "c": 3,
}

# Method 1: Using the | operator (Python 3.9+)
try:
    merged_pipe = dict1 | dict2
    print(f"dict1 | dict2: {merged_pipe}")
except Exception as e:
    print(f"Error with | operator: {e}")

# Method 2: Using the update method
merged_update = dict1.copy()
merged_update.update(dict2)
print(f"dict1.copy().update(dict2): {merged_update}")

# Method 3: Using union_recursive
merged_recursive = union_recursive(dict1, dict2)
print(f"union_recursive(dict1, dict2): {merged_recursive}")

# Compare the results
print("\nComparing the nested dictionary 'b':")
try:
    print(f"dict1 | dict2: {merged_pipe['b']}")
except Exception as e:
    print(f"Error with | operator: {e}")
print(f"dict1.copy().update(dict2): {merged_update['b']}")
print(f"union_recursive(dict1, dict2): {merged_recursive['b']}")
````

As you can see, both the `|` operator and the `update` method simply replace the entire nested dictionary at key `"b"` with the one from `dict2`, losing the value of `"x"` from `dict1`. In contrast, `union_recursive` merges the nested dictionaries, preserving the value of `"x"` from `dict1` while updating the value of `"y"` and adding the new key `"z"` from `dict2`.

### Preserving the Original Dictionaries

One important feature of `union_recursive` is that it creates a new dictionary without modifying the original dictionaries:

````
# Create two dictionaries
dict1 = {"a": 1, "b": {"x": 10}}
dict2 = {"b": {"y": 20}, "c": 3}

# Union them recursively
result = union_recursive(dict1, dict2)

# Modify the result
result["a"] = 100
result["b"]["x"] = 1000

# Check if the original dictionaries were modified
print(f"Original dict1: {dict1}")
print(f"Original dict2: {dict2}")
print(f"Modified result: {result}")
````

As you can see, modifying the result does not affect the original dictionaries. This is because `union_recursive` creates a deep copy of the first dictionary before updating it with the second dictionary.

## Module Interaction

The `unionrecursive` module interacts with other modules in the baseobjects package, particularly the `updaterecursive` module. It uses the `update_recursive` function from the `updaterecursive` module to perform the actual recursive update.

Let's see how this interaction works:

````
from baseobjects.operations.updaterecursive import update_recursive
from copy import deepcopy

# Create two dictionaries
dict1 = {"a": 1, "b": {"x": 10}}
dict2 = {"b": {"y": 20}, "c": 3}

# Method 1: Using union_recursive
result1 = union_recursive(dict1, dict2)

# Method 2: Using update_recursive with deepcopy
result2 = update_recursive(deepcopy(dict1), dict2)

# Compare the results
print(f"union_recursive(dict1, dict2): {result1}")
print(f"update_recursive(deepcopy(dict1), dict2): {result2}")
print(f"Are the results equal? {result1 == result2}")
````

As you can see, `union_recursive` is essentially a convenience function that combines `deepcopy` and `update_recursive`. It creates a deep copy of the first dictionary and then recursively updates it with the second dictionary.

## Advanced Features

### Working with Different Mapping Types

The `union_recursive` function works with any mapping type, not just dictionaries. Let's see how it works with other mapping types:

````
from collections import OrderedDict, defaultdict

# Create dictionaries of different types
regular_dict = {"a": 1, "b": {"x": 10}}
ordered_dict = OrderedDict([("b", {"y": 20}), ("c", 3)])
default_dict = defaultdict(int, {"d": 4, "e": 5})

# Union them recursively
result1 = union_recursive(regular_dict, ordered_dict)
result2 = union_recursive(result1, default_dict)

# Display the results
print(f"regular_dict: {regular_dict}")
print(f"ordered_dict: {ordered_dict}")
print(f"default_dict: {default_dict}")
print(f"union_recursive(regular_dict, ordered_dict): {result1}")
print(f"union_recursive(result1, default_dict): {result2}")

# Check the types
print(f"\nType of regular_dict: {type(regular_dict)}")
print(f"Type of ordered_dict: {type(ordered_dict)}")
print(f"Type of default_dict: {type(default_dict)}")
print(f"Type of result1: {type(result1)}")
print(f"Type of result2: {type(result2)}")
````

As you can see, the `union_recursive` function works with different mapping types. However, note that the type of the result is always the same as the type of the first argument. This is because `union_recursive` creates a deep copy of the first argument and then updates it with the second argument.

### Handling Non-Mapping Values

The `union_recursive` function only recursively merges values that are mappings. For non-mapping values, it simply replaces the value from the first dictionary with the value from the second dictionary:

````
# Create two dictionaries with various value types
dict1 = {
    "a": 1,
    "b": [1, 2, 3],
    "c": {"x": 10},
    "d": (4, 5, 6),
    "e": {1, 2, 3},
}

dict2 = {
    "a": "one",
    "b": [4, 5, 6],
    "c": {"y": 20},
    "d": {"p": 100},
    "e": 5,
}

# Union them recursively
result = union_recursive(dict1, dict2)

# Display the results
print(f"dict1: {dict1}")
print(f"dict2: {dict2}")
print(f"union_recursive(dict1, dict2): {result}")
````

As you can see, for keys `"a"`, `"b"`, and `"e"`, the values from `dict2` completely replace the values from `dict1`. For key `"c"`, which has a dictionary value in both dictionaries, the function recursively merges the dictionaries. For key `"d"`, which has a tuple value in `dict1` and a dictionary value in `dict2`, the value from `dict2` completely replaces the value from `dict1`.

## Examples

Let's explore some practical examples of using the `union_recursive` function.

### Example 1: Merging Configuration Settings

A common use case for recursive union is merging configuration settings, where you have default settings and user-specific settings:

````
# Default configuration
default_config = {
    "app": {
        "name": "MyApp",
        "version": "1.0.0",
    },
    "database": {
        "host": "localhost",
        "port": 5432,
        "username": "admin",
        "password": "admin",
        "settings": {
            "timeout": 30,
            "max_connections": 100,
        },
    },
    "logging": {
        "level": "INFO",
        "format": "%(asctime)s - %(name)s - %(levelname)s - %(message)s",
    },
}

# User-specific configuration
user_config = {
    "app": {
        "version": "1.0.1",  # User has a newer version
    },
    "database": {
        "host": "db.example.com",  # User has a different database host
        "username": "user",  # User has a different username
        "password": "password",  # User has a different password
        "settings": {
            "timeout": 60,  # User has a different timeout
        },
    },
}

# Merge the configurations
merged_config = union_recursive(default_config, user_config)

# Display the results
import json

print("Default configuration:")
print(json.dumps(default_config, indent=2))
print("\nUser-specific configuration:")
print(json.dumps(user_config, indent=2))
print("\nMerged configuration:")
print(json.dumps(merged_config, indent=2))
````

As you can see, the merged configuration contains all the settings from the default configuration, with the user-specific settings overriding the default settings where applicable. The nested settings are merged recursively, so the `max_connections` setting from the default configuration is preserved in the merged configuration.

### Example 2: Merging API Responses

Another use case is merging API responses, where you might receive partial updates to a resource:

````
# Initial resource state
resource = {
    "id": 123,
    "name": "Example Resource",
    "metadata": {
        "created_at": "2023-01-01T00:00:00Z",
        "updated_at": "2023-01-01T00:00:00Z",
        "tags": ["example", "resource"],
    },
    "properties": {
        "color": "blue",
        "size": "medium",
        "features": {
            "feature1": True,
            "feature2": False,
        },
    },
}

# Partial update from API
update = {
    "metadata": {
        "updated_at": "2023-07-15T12:30:00Z",
        "tags": ["example", "resource", "updated"],
    },
    "properties": {
        "color": "red",
        "features": {
            "feature2": True,
            "feature3": True,
        },
    },
}

# Merge the update into the resource
updated_resource = union_recursive(resource, update)

# Display the results
print("Initial resource state:")
print(json.dumps(resource, indent=2))
print("\nPartial update from API:")
print(json.dumps(update, indent=2))
print("\nUpdated resource:")
print(json.dumps(updated_resource, indent=2))
````

As you can see, the updated resource contains all the properties from the initial resource, with the properties from the update overriding the initial properties where applicable. The nested properties are merged recursively, so the `feature1` property from the initial resource is preserved in the updated resource.

A third use case is merging user preferences, where you might have global preferences and user-specific preferences:

````
# Global preferences
global_prefs = {
    "theme": {
        "mode": "light",
        "colors": {
            "primary": "#007bff",
            "secondary": "#6c757d",
            "success": "#28a745",
            "danger": "#dc3545",
        },
        "fonts": {
            "main": "Arial",
            "headings": "Helvetica",
            "code": "Courier New",
        },
    },
    "editor": {
        "indentation": 4,
        "word_wrap": True,
        "auto_save": True,
        "keybindings": {
            "save": "Ctrl+S",
            "find": "Ctrl+F",
            "replace": "Ctrl+H",
        },
    },
    "notifications": {
        "enabled": True,
        "sound": True,
        "desktop": True,
    },
}

# User-specific preferences
user_prefs = {
    "theme": {
        "mode": "dark",
        "colors": {
            "primary": "#0066cc",
            "danger": "#ff0000",
        },
    },
    "editor": {
        "word_wrap": False,
        "keybindings": {
            "save": "Cmd+S",  # User is on Mac
        },
    },
    "notifications": {
        "sound": False,
    },
}

# Merge the preferences
merged_prefs = union_recursive(global_prefs, user_prefs)

# Display the results
print("Global preferences:")
print(json.dumps(global_prefs, indent=2))
print("\nUser-specific preferences:")
print(json.dumps(user_prefs, indent=2))
print("\nMerged preferences:")
print(json.dumps(merged_prefs, indent=2))
````

```python
def union_recursive(d: Mapping, other: Mapping) -> Mapping:
    """Unions a mapping object and its contained mappings within another mapping.

        A new mapping as the union of the two mappings recursively.
    """
```

## Troubleshooting / FAQs

### Q: What happens if the same key exists in both mappings but with different value types?

A: If the same key exists in both mappings but with different value types, the value from the second mapping will completely replace the value from the first mapping. The function only recursively merges values that are both mappings.

### Q: Does `union_recursive` modify the original mappings?

A: No, `union_recursive` creates a new mapping without modifying the original mappings. It creates a deep copy of the first mapping and then recursively updates it with the second mapping.

### Q: What is the difference between `union_recursive` and `update_recursive`?

A: The main difference is that `union_recursive` creates a new mapping without modifying the original mappings, while `update_recursive` modifies the first mapping in place. `union_recursive` is essentially a convenience function that combines `deepcopy` and `update_recursive`.

### Q: What types of mappings can I use with `union_recursive`?

A: You can use any mapping type that implements the `Mapping` interface, such as `dict`, `OrderedDict`, `defaultdict`, and custom mapping types. However, note that the type of the result is always the same as the type of the first argument.
