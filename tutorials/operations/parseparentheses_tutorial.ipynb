# Using the ParseParentheses Module in baseobjects

## Introduction

The `parseparentheses` module provides functionality for parsing expressions with parentheses and returning a nested list of extracted elements. This is useful for working with structured text data, parsing command-line arguments, or implementing simple parsers for domain-specific languages.

The module contains a single function, `parse_parentheses`, which uses the `singlekwargdispatch` decorator to handle different input types (str, bytes, bytearray). The function parses expressions with parentheses into a structured list format, where each open parenthesis '(' denotes the start of a nested list, and each close parenthesis ')' indicates the end of the currently active list.

- Working with nested structures and filtering options

## Importing the Module

````
from baseobjects.operations.parseparentheses import parse_parentheses
````

## Core Functionality

The `parse_parentheses` function parses expressions with parentheses and returns a nested list of extracted elements. Let's explore its basic functionality.

### Basic Usage

Let's start with a simple example to see how the function works:

````
# Parse a simple expression with parentheses
expression = "a (b c) d"
result = parse_parentheses(expression)
print(f"Expression: {expression}")
print(f"Parsed result: {result}")
````

In this example, the function returns a list containing the elements 'a', ['b', 'c'], and 'd'. The elements 'b' and 'c' are nested within a sublist because they were enclosed in parentheses in the original expression.

Let's try a more complex example with nested parentheses:

````
# Parse a nested expression
nested_expression = "a (b (c d) e) f"
nested_result = parse_parentheses(nested_expression)
print(f"Expression: {nested_expression}")
print(f"Parsed result: {nested_result}")
````

Here, the function returns a list with elements 'a', a nested list ['b', ['c', 'd'], 'e'], and 'f'. The elements 'c' and 'd' are nested within a sublist, which is itself nested within another sublist.

### Different Input Types

The `parse_parentheses` function supports different input types. Let's see how it handles string, bytes, and bytearray inputs:

````
# String input (default)
string_expression = "a (b c) d"
string_result = parse_parentheses(string_expression)
print(f"String result: {string_result}")

# Bytes input
bytes_expression = b"a (b c) d"
bytes_result = parse_parentheses(bytes_expression)
print(f"Bytes result: {bytes_result}")

# Bytearray input
bytearray_expression = bytearray(b"a (b c) d")
bytearray_result = parse_parentheses(bytearray_expression)
print(f"Bytearray result: {bytearray_result}")
````

### Filtering Elements

The `parse_parentheses` function allows you to filter elements using the `include` and `exclude` parameters:

````
# Filter elements using include
expression = "a (b c) d e f"
include_result = parse_parentheses(expression, include={"a", "c", "e"})
print(f"Expression: {expression}")
print(f"Include only 'a', 'c', 'e': {include_result}")

# Filter elements using exclude
exclude_result = parse_parentheses(expression, exclude={"a", "c", "e"})
print(f"Exclude 'a', 'c', 'e': {exclude_result}")

# Combine include and exclude
combined_result = parse_parentheses(expression, include={"a", "b", "c", "d"}, exclude={"b", "d"})
print(f"Include 'a', 'b', 'c', 'd' but exclude 'b', 'd': {combined_result}")
````

### Casting Elements

The `parse_parentheses` function allows you to cast elements to a specific type using the `cast` parameter:

````
# Cast elements to uppercase
def to_uppercase(s):
    return s.upper() if isinstance(s, str) else s

expression = "a (b c) d"
uppercase_result = parse_parentheses(expression, cast=to_uppercase)
print(f"Expression: {expression}")
print(f"Cast to uppercase: {uppercase_result}")

# Cast numeric strings to integers
def to_int_if_numeric(s):
    try:
        return int(s)
    except (ValueError, TypeError):
        return s

numeric_expression = "1 (2 3) 4"
numeric_result = parse_parentheses(numeric_expression, cast=to_int_if_numeric)
print(f"Expression: {numeric_expression}")
print(f"Cast to integers: {numeric_result}")
````

## Advanced Features

### Handling Quoted Strings

The `parse_parentheses` function can handle quoted strings within expressions. Both single and double quotes are supported:

````
# Expression with quoted strings
quoted_expression = 'a ("quoted string") (\'another quoted string\') c'
quoted_result = parse_parentheses(quoted_expression)
print(f"Expression: {quoted_expression}")
print(f"Parsed result: {quoted_result}")

# Expression with quotes and nested parentheses
complex_quoted = 'command (sub1 "quoted (not a paren)") (sub2 (nested "still quoted"))'
complex_result = parse_parentheses(complex_quoted)
print(f"Expression: {complex_quoted}")
print(f"Parsed result: {complex_result}")
````

### Handling Unbalanced Parentheses

The `parse_parentheses` function raises a `ValueError` if the parentheses in the input expression are unbalanced:

````
# Unbalanced parentheses (missing closing parenthesis)
try:
    unbalanced1 = "a (b c"
    parse_parentheses(unbalanced1)
except ValueError as e:
    print(f"Error for '{unbalanced1}': {e}")

# Unbalanced parentheses (missing opening parenthesis)
try:
    unbalanced2 = "a b c)"
    parse_parentheses(unbalanced2)
except ValueError as e:
    print(f"Error for '{unbalanced2}': {e}")
````

### Custom Parsing with Regular Expressions

The `parseparentheses` module uses regular expressions internally to parse expressions. The regular expressions are defined as constants in the module:

````
# Import the regular expressions from the module
from baseobjects.operations.parseparentheses import r_expression, rb_expression

print(f"String expression pattern: {r_expression}")
print(f"Bytes expression pattern: {rb_expression}")

# Create a custom parser using the regular expressions
import re

def custom_parse(expression):
    """A simplified version of parse_parentheses for demonstration."""
    tokens = []
    for match in re.finditer(r_expression, expression.strip()):
        token = match.group(0)
        if token not in ["(", ")"]:
            tokens.append(token.strip())
    return tokens

# Test the custom parser
custom_result = custom_parse("a (b c) d")
print(f"Custom parser result: {custom_result}")
````

## Examples

Let's explore some practical examples of using the `parse_parentheses` function.

### Example 1: Parsing Command-Line Arguments

You can use the `parse_parentheses` function to parse command-line arguments with a hierarchical structure:

````
def parse_command(command_line):
    """Parse a command line with hierarchical arguments."""
    parsed = parse_parentheses(command_line)
    
    # Extract the command and arguments
    command = parsed[0] if parsed else ""
    args = []
    kwargs = {}
    
    for arg in parsed[1:]:
        if isinstance(arg, list) and len(arg) >= 2:
            # Treat nested lists as keyword arguments
            key = arg[0]
            value = arg[1:] if len(arg) > 2 else arg[1]
            kwargs[key] = value
        else:
            # Treat simple values as positional arguments
            args.append(arg)
    
    return {
        "command": command,
        "args": args,
        "kwargs": kwargs
    }

# Test with some example command lines
commands = [
    "find (path /home/user) (name *.txt) (type f)",
    "git commit (message \"Initial commit\") (author \"John Doe\")",
    "process data.csv (columns id name age) (filter (age > 18))"
]

for cmd in commands:
    parsed_cmd = parse_command(cmd)
    print(f"Command: {cmd}")
    print(f"  Command name: {parsed_cmd['command']}")
    print(f"  Positional args: {parsed_cmd['args']}")
    print(f"  Keyword args: {parsed_cmd['kwargs']}")
    print()
````

### Example 2: Parsing a Simple Expression Language

You can use the `parse_parentheses` function to parse a simple expression language:

````
def evaluate_expression(expression):
    """Evaluate a simple expression language."""
    parsed = parse_parentheses(expression)
    
    def eval_node(node):
        if not isinstance(node, list):
            # Try to convert to number if possible
            try:
                return float(node) if '.' in node else int(node)
            except (ValueError, TypeError):
                return node
        
        if not node:
            return None
        
        operator = node[0]
        operands = [eval_node(operand) for operand in node[1:]]
        
        if operator == '+':
            return sum(operands)
        elif operator == '-':
            return operands[0] - sum(operands[1:])
        elif operator == '*':
            result = 1
            for operand in operands:
                result *= operand
            return result
        elif operator == '/':
            result = operands[0]
            for operand in operands[1:]:
                result /= operand
            return result
        elif operator == 'concat':
            return ''.join(str(operand) for operand in operands)
        elif operator == 'list':
            return operands
        else:
            return [operator] + operands
    
    return eval_node(parsed[0] if len(parsed) == 1 and isinstance(parsed[0], list) else parsed)

# Test with some example expressions
expressions = [
    "(+ 1 2 3)",
    "(- 10 2 3)",
    "(* 2 3 4)",
    "(/ 24 2 3)",
    "(concat \"Hello\" \", \" \"World!\")",
    "(list 1 2 3 4)",
    "(+ 1 (* 2 3) (- 10 5))"
]

for expr in expressions:
    result = evaluate_expression(expr)
    print(f"Expression: {expr}")
    print(f"Result: {result}")
    print(f"Type: {type(result)}")
    print()
````

### Example 3: Parsing Nested Data Structures

You can use the `parse_parentheses` function to parse nested data structures like a simple JSON-like format:

````
def parse_data_structure(text):
    """Parse a simple data structure format."""
    parsed = parse_parentheses(text)
    
    def convert_node(node):
        if not isinstance(node, list):
            # Try to convert to appropriate type
            if node.lower() == 'true':
                return True
            elif node.lower() == 'false':
                return False
            elif node.lower() == 'null':
                return None
            try:
                if '.' in node:
                    return float(node)
                else:
                    return int(node)
            except (ValueError, TypeError):
                # Remove quotes if present
                if (node.startswith('"') and node.endswith('"')) or (node.startswith("'") and node.endswith("'")):
                    return node[1:-1]
                return node
        
        if not node:
            return []
        
        if node[0] == 'object':
            # Convert to dictionary
            result = {}
            for i in range(1, len(node), 2):
                if i + 1 < len(node):
                    key = convert_node(node[i])
                    value = convert_node(node[i + 1])
                    result[key] = value
            return result
        elif node[0] == 'array':
            # Convert to list
            return [convert_node(item) for item in node[1:]]
        else:
            # Unknown structure, return as is
            return [convert_node(item) for item in node]
    
    return convert_node(parsed[0] if len(parsed) == 1 and isinstance(parsed[0], list) else parsed)

# Test with some example data structures
data_structures = [
    "(object name \"John\" age 30 active true)",
    "(array 1 2 3 4 5)",
    "(object name \"Alice\" contact (object email \"alice@example.com\" phone \"123-456-7890\") skills (array \"Python\" \"JavaScript\" \"SQL\"))"
]

for data in data_structures:
    parsed_data = parse_data_structure(data)
    print(f"Data: {data}")
    print(f"Parsed: {parsed_data}")
    print(f"Type: {type(parsed_data)}")
    print()
````

## API Highlights

The `parseparentheses` module provides a single function:

### parse_parentheses

```python
@singlekwargdispatch
def parse_parentheses(
    expression: str | bytes | bytearray,
    include: set | None = None,
    exclude: set | None = None,
    cast: Callable = _no_cast,
) -> list[...]:
    """Parses expressions with parentheses and returns a nested list of extracted elements.

    This function uses a single-dispatch mechanism to handle parsing for various data types. If the type of the provided
    expression is not supported by any registered dispatch function, a ValueError is raised. Optional filtering can be
    applied by specifying `include` or `exclude` sets to control which elements are included or excluded. An optional
    casting function can be applied to transform the extracted elements.

    Args:
        expression: The input expression containing parentheses to parse.
        include: A set of elements to include in the output. Defaults to None.
        exclude: A set of elements to exclude from the output. Defaults to None.
        cast: A function to apply to each extracted element for custom transformations. Defaults to `_no_cast`.

    Returns:
        A nest list of parsed and optionally filtered and transformed elements.

    Raises:
        ValueError: If the type of the input expression is unsupported.
    """
```

Parameters:
- `expression`: The input expression containing parentheses to parse. Can be a str, bytes, or bytearray.
- `include`: A set of elements to include in the output. If provided, only elements in this set will be included. Defaults to None (include all).
- `exclude`: A set of elements to exclude from the output. If provided, elements in this set will be excluded. Defaults to None (exclude none).
- `cast`: A function to apply to each extracted element for custom transformations. Defaults to a function that returns the element unchanged.

Returns:
- A nested list of parsed and optionally filtered and transformed elements.

Raises:
- ValueError: If the parentheses in the input expression are unbalanced or if the type of the input expression is unsupported.

## Troubleshooting / FAQs

### Q: Why does my expression parse differently than expected?

A: There are a few possible reasons:
1. Quoted strings: The function treats text in quotes as a single element, even if it contains parentheses.
2. Whitespace: The function strips whitespace from elements by default.
3. Filtering: If you're using the `include` or `exclude` parameters, certain elements might be filtered out.

### Q: How does the function handle nested parentheses?

A: The function creates a nested list structure that mirrors the nesting of parentheses in the input expression. Each open parenthesis '(' starts a new nested list, and each close parenthesis ')' ends the current list.

### Q: Can I parse other types of brackets, like square brackets or curly braces?

A: No, the function is specifically designed to parse parentheses. If you need to parse other types of brackets, you would need to modify the function or create a new one based on the same approach.

### Q: How does the `cast` parameter work?

A: The `cast` parameter accepts a function that will be applied to each extracted element. This allows you to transform elements as they are parsed. For example, you could use it to convert numeric strings to integers or to normalize text.
