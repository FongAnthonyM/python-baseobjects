# Using the UpdateRecursive Module in baseobjects

## Introduction

The `updaterecursive` module provides functionality for recursively updating mapping objects and their contained mappings. This is useful when you need to modify nested dictionaries or other mapping types while preserving the hierarchical structure.

The module contains two functions:
- `update_recursive`: The main function that updates a mapping object and its contained mappings based on another mapping
- `_update_recursive`: A helper function used internally by `update_recursive`

Unlike the standard dictionary update method, which only updates at the top level, `update_recursive` will recursively update nested mappings as well. This is particularly useful when working with complex, nested data structures like configuration settings, API responses, or user preferences.

- Understanding the purpose and functionality of the `updaterecursive` module
- Using the `update_recursive` function with different mapping types
- Comparing recursive update with standard dictionary operations
- Practical examples and use cases

- Familiarity with dictionary operations like update and merge

## Core Functionality

The `update_recursive` function updates a mapping object and its contained mappings based on another mapping. Let's explore its basic functionality.

### Basic Usage

Let's start with a simple example to see how the function works:

````
# Create two dictionaries
dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"b": 20, "c": 30, "d": 40}

# Make a copy of dict1 to preserve the original
dict1_copy = dict1.copy()

# Update dict1_copy recursively with dict2
result = update_recursive(dict1_copy, dict2)

# Display the results
print(f"Original dict1: {dict1}")
print(f"dict2: {dict2}")
print(f"update_recursive(dict1_copy, dict2): {result}")
print(f"dict1_copy after update: {dict1_copy}")
print(f"Is result the same object as dict1_copy? {result is dict1_copy}")
````

In this example, the function updates `dict1_copy` with the values from `dict2` and returns the updated dictionary. For keys that exist in both dictionaries, the values from the second dictionary (`dict2`) take precedence. Note that `update_recursive` modifies the first dictionary in place and returns it, so `result` and `dict1_copy` are the same object.

Now, let's try with nested dictionaries:

````
# Create two nested dictionaries
nested_dict1 = {
    "a": 1,
    "b": {
        "x": 10,
        "y": 20,
    },
    "c": {
        "p": 100,
        "q": 200,
    },
}

nested_dict2 = {
    "b": {
        "y": 30,
        "z": 40,
    },
    "c": 300,
    "d": 400,
}

# Make a copy of nested_dict1 to preserve the original
nested_dict1_copy = nested_dict1.copy()

# Update nested_dict1_copy recursively with nested_dict2
nested_result = update_recursive(nested_dict1_copy, nested_dict2)

# Display the results
print(f"Original nested_dict1: {nested_dict1}")
print(f"nested_dict2: {nested_dict2}")
print(f"update_recursive(nested_dict1_copy, nested_dict2): {nested_result}")
print(f"nested_dict1_copy after update: {nested_dict1_copy}")
````

In this example, the function recursively updates the nested dictionaries. For the key `"b"`, which exists in both dictionaries and has a dictionary value in both, the function recursively updates the nested dictionary. For the key `"c"`, which exists in both dictionaries but has a dictionary value in `nested_dict1` and a non-dictionary value in `nested_dict2`, the value from `nested_dict2` completely replaces the value from `nested_dict1`.

### Comparison with Standard Dictionary Operations

Let's compare `update_recursive` with standard dictionary operations:

````
# Create two nested dictionaries
dict1 = {
    "a": 1,
    "b": {
        "x": 10,
        "y": 20,
    },
}

dict2 = {
    "b": {
        "y": 30,
        "z": 40,
    },
    "c": 3,
}

# Method 1: Using the update method
dict1_update = dict1.copy()
dict1_update.update(dict2)
print(f"dict1.copy().update(dict2): {dict1_update}")

# Method 2: Using update_recursive
dict1_recursive = dict1.copy()
update_recursive(dict1_recursive, dict2)
print(f"update_recursive(dict1.copy(), dict2): {dict1_recursive}")

# Compare the results
print("\nComparing the nested dictionary 'b':")
print(f"dict1.copy().update(dict2): {dict1_update['b']}")
print(f"update_recursive(dict1.copy(), dict2): {dict1_recursive['b']}")
````

As you can see, the standard `update` method simply replaces the entire nested dictionary at key `"b"` with the one from `dict2`, losing the value of `"x"` from `dict1`. In contrast, `update_recursive` recursively updates the nested dictionary, preserving the value of `"x"` from `dict1` while updating the value of `"y"` and adding the new key `"z"` from `dict2`.

### Working with Different Input Types

The `update_recursive` function can accept the updates as either a mapping or an iterable of key-value pairs:

````
# Create a dictionary
dict1 = {"a": 1, "b": {"x": 10, "y": 20}}

# Update with a mapping
dict1_copy1 = dict1.copy()
update_recursive(dict1_copy1, {"b": {"y": 30, "z": 40}, "c": 3})
print(f"Update with a mapping: {dict1_copy1}")

# Update with an iterable of key-value pairs
dict1_copy2 = dict1.copy()
update_recursive(dict1_copy2, [("b", {"y": 30, "z": 40}), ("c", 3)])
print(f"Update with an iterable: {dict1_copy2}")

# Update with a generator expression
dict1_copy3 = dict1.copy()
update_recursive(dict1_copy3, ((k, v) for k, v in [("b", {"y": 30, "z": 40}), ("c", 3)]))
print(f"Update with a generator: {dict1_copy3}")
````

As you can see, the function works with different types of inputs for the updates parameter, including mappings, iterables of key-value pairs, and generator expressions.

## Module Interaction

The `updaterecursive` module is used by other modules in the baseobjects package, particularly the `unionrecursive` module. The `union_recursive` function from the `unionrecursive` module uses `update_recursive` to perform the actual recursive update.

Let's see how this interaction works:

````
from baseobjects.operations.unionrecursive import union_recursive
from copy import deepcopy

# Create two dictionaries
dict1 = {"a": 1, "b": {"x": 10}}
dict2 = {"b": {"y": 20}, "c": 3}

# Method 1: Using union_recursive
result1 = union_recursive(dict1, dict2)

# Method 2: Using update_recursive with deepcopy
result2 = update_recursive(deepcopy(dict1), dict2)

# Compare the results
print(f"union_recursive(dict1, dict2): {result1}")
print(f"update_recursive(deepcopy(dict1), dict2): {result2}")
print(f"Are the results equal? {result1 == result2}")
````

As you can see, `union_recursive` is essentially a convenience function that combines `deepcopy` and `update_recursive`. It creates a deep copy of the first dictionary and then recursively updates it with the second dictionary, returning the result without modifying the original dictionaries.

## Advanced Features

### Working with Different Mapping Types

The `update_recursive` function works with any mapping type, not just dictionaries. Let's see how it works with other mapping types:

````
from collections import OrderedDict, defaultdict

# Create dictionaries of different types
regular_dict = {"a": 1, "b": {"x": 10}}
ordered_dict = OrderedDict([("b", {"y": 20}), ("c", 3)])
default_dict = defaultdict(int, {"d": 4, "e": 5})

# Update regular_dict with ordered_dict
regular_dict_copy = regular_dict.copy()
update_recursive(regular_dict_copy, ordered_dict)
print(f"regular_dict updated with ordered_dict: {regular_dict_copy}")

# Update ordered_dict with default_dict
ordered_dict_copy = ordered_dict.copy()
update_recursive(ordered_dict_copy, default_dict)
print(f"ordered_dict updated with default_dict: {ordered_dict_copy}")

# Update default_dict with regular_dict
default_dict_copy = default_dict.copy()
update_recursive(default_dict_copy, regular_dict)
print(f"default_dict updated with regular_dict: {default_dict_copy}")

# Check the types
print(f"\nType of regular_dict_copy: {type(regular_dict_copy)}")
print(f"Type of ordered_dict_copy: {type(ordered_dict_copy)}")
print(f"Type of default_dict_copy: {type(default_dict_copy)}")
````

As you can see, the `update_recursive` function works with different mapping types. The type of the result is always the same as the type of the first argument, since the function modifies the first argument in place.

### Handling Non-Mapping Values

The `update_recursive` function only recursively updates values that are mappings. For non-mapping values, it simply replaces the value from the first dictionary with the value from the second dictionary:

````
# Create two dictionaries with various value types
dict1 = {
    "a": 1,
    "b": [1, 2, 3],
    "c": {"x": 10},
    "d": (4, 5, 6),
    "e": {1, 2, 3},
}

dict2 = {
    "a": "one",
    "b": [4, 5, 6],
    "c": {"y": 20},
    "d": {"p": 100},
    "e": 5,
}

# Update dict1 with dict2
dict1_copy = dict1.copy()
update_recursive(dict1_copy, dict2)

# Display the results
print(f"Original dict1: {dict1}")
print(f"dict2: {dict2}")
print(f"dict1_copy after update: {dict1_copy}")
````

As you can see, for keys `"a"`, `"b"`, and `"e"`, the values from `dict2` completely replace the values from `dict1`. For key `"c"`, which has a dictionary value in both dictionaries, the function recursively updates the dictionary. For key `"d"`, which has a tuple value in `dict1` and a dictionary value in `dict2`, the value from `dict2` completely replaces the value from `dict1`.

### Chaining Updates

you can chain multiple updates together by calling `update_recursive` multiple times:

````
# Create three dictionaries
dict1 = {"a": 1, "b": {"x": 10}}
dict2 = {"b": {"y": 20}, "c": 3}
dict3 = {"c": 30, "d": {"p": 100}}

# Chain updates
result = dict1.copy()
update_recursive(result, dict2)
update_recursive(result, dict3)

# Display the results
print(f"dict1: {dict1}")
print(f"dict2: {dict2}")
print(f"dict3: {dict3}")
print(f"result after chained updates: {result}")
````

As you can see, you can chain multiple updates together by calling `update_recursive` multiple times on the same dictionary. Each call updates the dictionary in place with the values from the new dictionary.

## Examples

Let's explore some practical examples of using the `update_recursive` function.

### Example 1: Updating Configuration Settings

A common use case for recursive update is updating configuration settings, where you have default settings and want to apply user-specific settings:

````
# Default configuration
default_config = {
    "app": {
        "name": "MyApp",
        "version": "1.0.0",
    },
    "database": {
        "host": "localhost",
        "port": 5432,
        "username": "admin",
        "password": "admin",
        "settings": {
            "timeout": 30,
            "max_connections": 100,
        },
    },
    "logging": {
        "level": "INFO",
        "format": "%(asctime)s - %(name)s - %(levelname)s - %(message)s",
    },
}

# User-specific configuration
user_config = {
    "app": {
        "version": "1.0.1",  # User has a newer version
    },
    "database": {
        "host": "db.example.com",  # User has a different database host
        "username": "user",  # User has a different username
        "password": "password",  # User has a different password
        "settings": {
            "timeout": 60,  # User has a different timeout
        },
    },
}

# Update the default configuration with the user-specific configuration
config = default_config.copy()
update_recursive(config, user_config)

# Display the results
import json

print("Default configuration:")
print(json.dumps(default_config, indent=2))
print("\nUser-specific configuration:")
print(json.dumps(user_config, indent=2))
print("\nUpdated configuration:")
print(json.dumps(config, indent=2))
````

As you can see, the updated configuration contains all the settings from the default configuration, with the user-specific settings overriding the default settings where applicable. The nested settings are updated recursively, so the `max_connections` setting from the default configuration is preserved in the updated configuration.

### Example 2: Updating API Resources

Another use case is updating API resources, where you might receive partial updates to a resource:

````
# Initial resource state
resource = {
    "id": 123,
    "name": "Example Resource",
    "metadata": {
        "created_at": "2023-01-01T00:00:00Z",
        "updated_at": "2023-01-01T00:00:00Z",
        "tags": ["example", "resource"],
    },
    "properties": {
        "color": "blue",
        "size": "medium",
        "features": {
            "feature1": True,
            "feature2": False,
        },
    },
}

# Partial update from API
update = {
    "metadata": {
        "updated_at": "2023-07-15T12:30:00Z",
        "tags": ["example", "resource", "updated"],
    },
    "properties": {
        "color": "red",
        "features": {
            "feature2": True,
            "feature3": True,
        },
    },
}

# Update the resource
update_recursive(resource, update)

# Display the results
print("Updated resource:")
print(json.dumps(resource, indent=2))
````

As you can see, the resource has been updated with the values from the update. The nested properties are updated recursively, so the `feature1` property from the initial resource is preserved in the updated resource.

### Example 3: Implementing a Settings Manager

A third use case is implementing a settings manager that allows users to update their settings:

````
class SettingsManager:
    """A simple settings manager that supports recursive updates."""

    def __init__(self, default_settings=None) -> None:
        """Initialize the settings manager with default settings."""
        self.settings = default_settings or {}

    def update_settings(self, new_settings) -> None:
        """Update the settings recursively."""
        update_recursive(self.settings, new_settings)

    def get_settings(self):
        """Get the current settings."""
        return self.settings

    def get_setting(self, path, default=None):
        """Get a specific setting by path."""
        parts = path.split(".")
        current = self.settings

        for part in parts:
            if isinstance(current, dict) and part in current:
                current = current[part]
            else:
                return default

        return current


# Create a settings manager with default settings
manager = SettingsManager({
    "theme": {
        "mode": "light",
        "colors": {
            "primary": "#007bff",
            "secondary": "#6c757d",
        },
    },
    "editor": {
        "indentation": 4,
        "word_wrap": True,
    },
})

# Display the initial settings
print("Initial settings:")
print(json.dumps(manager.get_settings(), indent=2))

# Update the settings
manager.update_settings({
    "theme": {
        "mode": "dark",
        "colors": {
            "primary": "#0066cc",
        },
    },
    "editor": {
        "word_wrap": False,
    },
})

# Display the updated settings
print("\nUpdated settings:")
print(json.dumps(manager.get_settings(), indent=2))

# Get specific settings
print("\nSpecific settings:")
print(f"Theme mode: {manager.get_setting('theme.mode')}")
print(f"Primary color: {manager.get_setting('theme.colors.primary')}")
print(f"Secondary color: {manager.get_setting('theme.colors.secondary')}")
print(f"Indentation: {manager.get_setting('editor.indentation')}")
print(f"Word wrap: {manager.get_setting('editor.word_wrap')}")
print(f"Nonexistent setting: {manager.get_setting('nonexistent', 'default value')}")
````

As you can see, the `SettingsManager` class uses `update_recursive` to update the settings recursively. This allows users to update specific settings without having to provide the entire settings object.

## Troubleshooting / FAQs

### Q: What happens if the same key exists in both mappings but with different value types?

A: If the same key exists in both mappings but with different value types, the value from the second mapping will completely replace the value from the first mapping. The function only recursively updates values that are both mappings.

### Q: Does `update_recursive` modify the original mapping?

A: Yes, `update_recursive` modifies the first mapping in place and returns it. If you want to preserve the original mapping, you should make a copy of it before calling `update_recursive`.

### Q: What is the difference between `update_recursive` and `union_recursive`?

A: The main difference is that `update_recursive` modifies the first mapping in place, while `union_recursive` creates a new mapping without modifying the original mappings. `union_recursive` is essentially a convenience function that combines `deepcopy` and `update_recursive`.

### Q: What types of mappings can I use with `update_recursive`?

A: you can use any mapping type that implements the `Mapping` interface, such as `dict`, `OrderedDict`, `defaultdict`, and custom mapping types. The function will preserve the type of the first mapping.

### Q: Can I use `update_recursive` with non-mapping values?

A: The first argument to `update_recursive` must be a mapping, but the second argument can be either a mapping or an iterable of key-value pairs. For non-mapping values within the mappings, the function will simply replace the value from the first mapping with the value from the second mapping.
