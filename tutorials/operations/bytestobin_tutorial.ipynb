# Using the BytesToBin Module in baseobjects

## Introduction

The `bytestobin` module provides a function for converting bytes to a tuple of binary values. This is useful when you need to work with binary data at the bit level, such as when implementing low-level protocols, working with hardware interfaces, or performing bit-level operations.

The module contains a single function, `bytes_to_bin`, which takes a bytes object and returns a tuple of binary values (0s and 1s) representing the individual bits in the bytes. The function supports different byte orders and output types, making it flexible for various use cases.

## Importing the Module


````
from baseobjects.operations.bytestobin import bytes_to_bin
````

## Core Functionality

The `bytes_to_bin` function converts a bytes object to a tuple of binary values. Each bit in the input bytes is represented as a separate value in the output tuple.

### Basic Usage

Let's start with a simple example to see how the function works:


````
# Convert a single byte to binary
single_byte = bytes([0b10101010])  # Binary: 10101010
binary_tuple = bytes_to_bin(single_byte)
print(f"Input byte: {single_byte}")
print(f"Binary representation: {binary_tuple}")
print(f"As a string: {''.join(map(str, binary_tuple))}")
````

By default, the function returns a tuple of integers (0 or 1). Let's try with multiple bytes:


````
# Convert multiple bytes to binary
multi_bytes = bytes([0b11001100, 0b10101010])
binary_tuple = bytes_to_bin(multi_bytes)
print(f"Input bytes: {multi_bytes}")
print(f"Binary representation: {binary_tuple}")
print(f"As a string: {''.join(map(str, binary_tuple))}")
````

### Byte Order

The `bytes_to_bin` function supports different byte orders through the `byteorder` parameter. The default is "big" endian, but you can also use "little" endian:


````
# Compare big and little endian byte order
test_bytes = bytes([0b10000001, 0b01000010])

# Big endian (default)
big_endian = bytes_to_bin(test_bytes, byteorder="big")
print(f"Big endian: {''.join(map(str, big_endian))}")

# Little endian
little_endian = bytes_to_bin(test_bytes, byteorder="little")
print(f"Little endian: {''.join(map(str, little_endian))}")
````

### Output Type

The `bytes_to_bin` function allows you to specify the output type through the `out_type` parameter. By default, it returns integers (0 or 1), but you can also get booleans, strings, or any other type that can be constructed from a boolean:


````
# Different output types
test_byte = bytes([0b10101010])

# Integer output (default)
int_output = bytes_to_bin(test_byte, out_type=int)
print(f"Integer output: {int_output}")

# Boolean output
bool_output = bytes_to_bin(test_byte, out_type=bool)
print(f"Boolean output: {bool_output}")

# String output
str_output = bytes_to_bin(test_byte, out_type=str)
print(f"String output: {str_output}")

# Custom output type
class BitValue:
    def __init__(self, value):
        self.value = value
    def __repr__(self):
        return f"Bit({self.value})"

custom_output = bytes_to_bin(test_byte, out_type=BitValue)
print(f"Custom output: {custom_output}")
````

## Advanced Features

### Working with Different Data Sources

The `bytes_to_bin` function works with any bytes-like object. Here are some examples of different ways to create input data:


````
# From a bytes literal
bytes_literal = b'\x55\xAA'  # 01010101 10101010
print(f"From bytes literal: {''.join(map(str, bytes_to_bin(bytes_literal)))}")

# From a bytearray
byte_array = bytearray([0x55, 0xAA])  # 01010101 10101010
print(f"From bytearray: {''.join(map(str, bytes_to_bin(byte_array)))}")

# From a string encoded to bytes
string_bytes = "AB".encode('ascii')  # ASCII: A=65 (01000001), B=66 (01000010)
print(f"From string: {''.join(map(str, bytes_to_bin(string_bytes)))}")

# From an integer converted to bytes
int_bytes = (1234).to_bytes(2, byteorder='big')  # 1234 = 0x04D2 = 00000100 11010010
print(f"From integer: {''.join(map(str, bytes_to_bin(int_bytes)))}")
````

### Combining with Other Binary Operations

You can combine `bytes_to_bin` with other binary operations to perform more complex tasks:


````
# Count the number of 1 bits in a byte (population count)
def popcount(byte):
    binary = bytes_to_bin(bytes([byte]))
    return sum(binary)

print(f"Popcount of 0b10101010: {popcount(0b10101010)}")
print(f"Popcount of 0b11111111: {popcount(0b11111111)}")
print(f"Popcount of 0b00000000: {popcount(0b00000000)}")
````

## Examples

Let's explore some practical examples of using the `bytes_to_bin` function.

### Example 1: Parsing a Binary Protocol

Imagine we have a simple binary protocol where each bit in a byte has a specific meaning. We can use `bytes_to_bin` to parse this protocol:


````
def parse_status_byte(status_byte):
    """Parse a status byte where each bit has a specific meaning."""
    bits = bytes_to_bin(bytes([status_byte]))
    return {
        'power_on': bool(bits[0]),
        'error': bool(bits[1]),
        'busy': bool(bits[2]),
        'data_ready': bool(bits[3]),
        'reserved1': bool(bits[4]),
        'reserved2': bool(bits[5]),
        'mode_bit1': bool(bits[6]),
        'mode_bit2': bool(bits[7])
    }

# Example status bytes
status1 = 0b10000101  # Power on, busy, mode_bit2 on
status2 = 0b01001001  # Power on, data_ready, error

print("Status 1:")
for key, value in parse_status_byte(status1).items():
    print(f"  {key}: {value}")

print("\nStatus 2:")
for key, value in parse_status_byte(status2).items():
    print(f"  {key}: {value}")
````

### Example 2: Converting Between Binary Representations

We can use `bytes_to_bin` to convert between different binary representations:


````
def bytes_to_binary_string(bytes_obj):
    """Convert bytes to a binary string with spaces between bytes."""
    binary_tuple = bytes_to_bin(bytes_obj)
    binary_string = ''.join(map(str, binary_tuple))
    # Insert spaces every 8 bits
    return ' '.join(binary_string[i:i+8] for i in range(0, len(binary_string), 8))

# Convert various data types to binary strings
text = "Hello"
text_bytes = text.encode('ascii')
print(f"Text: '{text}'")
print(f"Binary: {bytes_to_binary_string(text_bytes)}")

number = 42
number_bytes = number.to_bytes(1, byteorder='big')
print(f"\nNumber: {number}")
print(f"Binary: {bytes_to_binary_string(number_bytes)}")

# Convert a binary string back to bytes
def binary_string_to_bytes(binary_string):
    """Convert a binary string to bytes."""
    # Remove spaces
    binary_string = binary_string.replace(' ', '')
    # Ensure the length is a multiple of 8
    if len(binary_string) % 8 != 0:
        binary_string = binary_string.zfill((len(binary_string) // 8 + 1) * 8)
    # Convert to bytes
    return bytes(int(binary_string[i:i+8], 2) for i in range(0, len(binary_string), 8))

binary_string = "01001000 01100101 01101100 01101100 01101111"  # "Hello"
print(f"\nBinary string: {binary_string}")
bytes_obj = binary_string_to_bytes(binary_string)
print(f"Bytes: {bytes_obj}")
print(f"Text: {bytes_obj.decode('ascii')}")
````

## API Highlights

The `bytestobin` module provides a single function:

### bytes_to_bin

```python
def bytes_to_bin(bytes_: bytes, byteorder: str = "big", out_type: type = int) -> tuple[Any]:
    """Converts bytes to a tuple of binary values.

    Args:
        bytes_: The bytes to convert to binary.
        byteorder: The byte order of bytes to use for the conversion, either 'little' or 'big'.
        out_type: The type to represent the binary values as, such as int, bool, or str,

    Returns:
        The tuple of binary values from the bytes.
    """
```

Parameters:
- `bytes_`: The bytes object to convert to binary values.
- `byteorder`: The byte order to use for the conversion, either 'little' or 'big'. Default is 'big'.
- `out_type`: The type to represent the binary values as, such as int, bool, or str. Default is int.

Returns:
- A tuple of binary values, where each value represents a bit in the input bytes.


## Troubleshooting / FAQs

### Q: What's the difference between 'big' and 'little' endian byte order?

A: The `byteorder` parameter determines the order in which bits are extracted from each byte:
- In 'big' endian mode (the default), bits are extracted from most significant bit (MSB) to least significant bit (LSB): 128, 64, 32, 16, 8, 4, 2, 1.
- In 'little' endian mode, bits are extracted in a different order: 8, 4, 2, 1, 128, 64, 32, 16.

This affects the order of bits in the output tuple.

### Q: Why does the function return a tuple instead of a list?

A: Tuples are immutable, which makes them more appropriate for representing fixed binary data. If you need a mutable structure, you can convert the result to a list: `list(bytes_to_bin(data))`.

### Q: How does the `out_type` parameter work?

A: The `out_type` parameter specifies the type constructor to apply to each bit value. The function first converts each bit to a boolean (True for 1, False for 0), then applies the specified type constructor. For example:
- `out_type=int` converts True to 1 and False to 0
- `out_type=bool` keeps the values as True and False
- `out_type=str` converts True to "True" and False to "False"

You can also use custom types as long as they can be constructed from a boolean value.
