# Using the MethodNames Module in baseobjects

## Introduction

The `methodnames` module provides utility functions for retrieving method names from Python objects. This is useful for introspection, debugging, and metaprogramming tasks where you need to examine the methods available on an object.

The module contains four main functions:
- `iter_method_names`: Creates an iterator over all method names of an object
- `iter_public_method_names`: Creates an iterator over public method names (those not starting with underscore)
- `get_method_names`: Returns a tuple of all method names
- `get_public_method_names`: Returns a tuple of all public method names

This tutorial will guide you through:
- Understanding the purpose and functionality of the `methodnames` module
- Using the different functions to retrieve method names
- Practical examples and use cases for method introspection

**Prerequisites:**
- Basic understanding of Python objects and methods
- Familiarity with Python's introspection capabilities
- Knowledge of Python iterators and generators

## Importing the Module

````
from baseobjects.operations.methodnames import (
    iter_method_names,
    iter_public_method_names,
    get_method_names,
    get_public_method_names,
)
````

## Core Functionality

The `methodnames` module provides functions for retrieving method names from Python objects. Let's explore its basic functionality.

### Basic Usage

Let's start with a simple example to see how the functions work:

````
# Create a simple class with various methods
class ExampleClass:
    def public_method(self) -> None:
        """A public method."""

    def another_public_method(self) -> None:
        """Another public method."""

    def _private_method(self) -> None:
        """A private method (by convention)."""

    def __special_method__(self):
        """A special method."""


# Create an instance of the class
example = ExampleClass()

# Get all method names
all_methods = get_method_names(example)
print("All methods:")
for method in all_methods:
    print(f"  {method}")

# Get public method names
public_methods = get_public_method_names(example)
print("\nPublic methods:")
for method in public_methods:
    print(f"  {method}")
````

### Using Iterators

The module also provides iterator versions of the functions, which can be more memory-efficient for large objects:

````
# Using iterators for method names
print("All methods (using iterator):")
for method in iter_method_names(example):
    print(f"  {method}")

print("\nPublic methods (using iterator):")
for method in iter_public_method_names(example):
    print(f"  {method}")
````

### Working with Built-in Types

The functions work with any Python object, including built-in types:

````
# Get method names from a string
string_methods = get_method_names("hello")
print(f"String has {len(string_methods)} methods")
print("First 10 string methods:")
for method in list(string_methods)[:10]:
    print(f"  {method}")

# Get method names from a list
list_methods = get_method_names([1, 2, 3])
print(f"\nList has {len(list_methods)} methods")
print("First 10 list methods:")
for method in list(list_methods)[:10]:
    print(f"  {method}")

# Get method names from a dictionary
dict_methods = get_method_names({"a": 1, "b": 2})
print(f"\nDictionary has {len(dict_methods)} methods")
print("First 10 dictionary methods:")
for method in list(dict_methods)[:10]:
    print(f"  {method}")
````

## Advanced Features

### Filtering Method Names

You can use the functions from the `methodnames` module along with Python's built-in filtering capabilities to create more specific method name filters:

````
# Get all methods from a list
list_obj = [1, 2, 3]
all_list_methods = get_method_names(list_obj)

# Filter for methods that modify the list in-place
modifying_methods = [
    method
    for method in all_list_methods
    if not method.startswith("__")
    and not method.startswith("_")
    and not method.startswith("count")
    and not method.startswith("index")
    and not method.startswith("copy")
]
print("List-modifying methods:")
for method in modifying_methods:
    print(f"  {method}")

# Filter for methods that don't modify the list (query methods)
query_methods = [method for method in all_list_methods if method in ["count", "index"]]
print("\nList query methods:")
for method in query_methods:
    print(f"  {method}")

# Filter for special methods (dunder methods)
special_methods = [method for method in all_list_methods if method.startswith("__") and method.endswith("__")]
print("\nList special methods (first 10):")
for method in special_methods[:10]:
    print(f"  {method}")
````

### Comparing Objects by Available Methods

You can use the functions to compare the methods available on different objects:

````
# Compare methods available on different types
string_methods = set(get_method_names("hello"))
list_methods = set(get_method_names([1, 2, 3]))
dict_methods = set(get_method_names({"a": 1}))

# Methods common to all three types
common_methods = string_methods.intersection(list_methods, dict_methods)
print(f"Methods common to strings, lists, and dictionaries ({len(common_methods)}):")
for method in sorted(common_methods)[:10]:  # Show first 10
    print(f"  {method}")

# Methods unique to strings
string_unique = string_methods - list_methods - dict_methods
print(f"\nMethods unique to strings ({len(string_unique)}):")
for method in sorted(string_unique)[:10]:  # Show first 10
    print(f"  {method}")

# Methods unique to lists
list_unique = list_methods - string_methods - dict_methods
print(f"\nMethods unique to lists ({len(list_unique)}):")
for method in sorted(list_unique)[:10]:  # Show first 10
    print(f"  {method}")

# Methods unique to dictionaries
dict_unique = dict_methods - string_methods - list_methods
print(f"\nMethods unique to dictionaries ({len(dict_unique)}):")
for method in sorted(dict_unique)[:10]:  # Show first 10
    print(f"  {method}")
````

## Examples

Let's explore some practical examples of using the `methodnames` module.

### Example 1: Creating a Method Directory

You can use the `methodnames` module to create a simple method directory for an object, categorizing methods by their purpose:

````
def create_method_directory(obj):
    """Create a categorized directory of methods for an object."""
    all_methods = get_method_names(obj)

    # Categorize methods
    special_methods = []
    private_methods = []
    public_methods = []

    for method in all_methods:
        if method.startswith("__") and method.endswith("__"):
            special_methods.append(method)
        elif method.startswith("_"):
            private_methods.append(method)
        else:
            public_methods.append(method)

    # Create the directory
    return {
        "special_methods": sorted(special_methods),
        "private_methods": sorted(private_methods),
        "public_methods": sorted(public_methods),
    }


# Create a method directory for a list
list_directory = create_method_directory([1, 2, 3])

# Display the directory
print("List Method Directory:")
print(f"  Special methods: {len(list_directory['special_methods'])}")
print(f"  Private methods: {len(list_directory['private_methods'])}")
print(f"  Public methods: {len(list_directory['public_methods'])}")

print("\nPublic methods:")
for method in list_directory["public_methods"]:
    print(f"  {method}")

print("\nFirst 5 special methods:")
for method in list_directory["special_methods"][:5]:
    print(f"  {method}")
````

### Example 2: Implementing a Simple Proxy Object

You can use the `methodnames` module to implement a simple proxy object that forwards method calls to another object:

````
class SimpleProxy:
    """A simple proxy that forwards method calls to another object."""

    def __init__(self, target) -> None:
        self._target = target

        # Get all public methods of the target
        public_methods = get_public_method_names(target)

        # Create proxy methods for each public method
        for method_name in public_methods:
            # Skip methods that we're already defining
            if method_name in ["__init__", "__getattr__"]:
                continue

            # Get the method from the target
            getattr(target, method_name)

            # Create a proxy method
            def make_proxy_method(name):
                def proxy_method(*args, **kwargs):
                    method = getattr(self._target, name)
                    result = method(*args, **kwargs)
                    print(f"Called {name}({args}, {kwargs}) -> {result}")
                    return result

                return proxy_method

            # Set the proxy method on this object
            setattr(self, method_name, make_proxy_method(method_name))

    def __getattr__(self, name):
        """Forward any other attribute access to the target."""
        return getattr(self._target, name)


# Create a proxy for a list
original_list = [1, 2, 3, 4, 5]
proxy_list = SimpleProxy(original_list)

# Use the proxy
print(f"Original list: {original_list}")
print(f"Proxy list: {proxy_list}")

# Call some methods on the proxy
proxy_list.append(6)
proxy_list.extend([7, 8])
proxy_list.remove(1)
print(f"After modifications: {proxy_list}")
````

### Example 3: Creating a Method Usage Tracker

You can use the `methodnames` module to create a decorator that tracks method usage:

````
def track_method_usage(cls):
    """Class decorator that tracks method usage."""
    original_methods = {}
    usage_counts = {}

    # Get all public methods
    for method_name in get_public_method_names(cls):
        # Skip special methods
        if method_name.startswith("__"):
            continue

        # Store the original method
        original_method = getattr(cls, method_name)
        original_methods[method_name] = original_method
        usage_counts[method_name] = 0

        # Create a tracking wrapper
        def make_wrapper(name, method):
            def wrapper(self, *args, **kwargs):
                usage_counts[name] += 1
                return method(self, *args, **kwargs)

            return wrapper

        # Replace the method with the wrapper
        setattr(cls, method_name, make_wrapper(method_name, original_method))

    # Add a method to get usage statistics
    def get_usage_stats(self):
        return dict(usage_counts)

    cls.get_usage_stats = get_usage_stats

    return cls


# Apply the decorator to a class
@track_method_usage
class Counter:
    def __init__(self) -> None:
        self.count = 0

    def increment(self):
        self.count += 1
        return self.count

    def decrement(self):
        self.count -= 1
        return self.count

    def reset(self):
        self.count = 0
        return self.count

    def get_count(self):
        return self.count


# Create an instance and use its methods
counter = Counter()
counter.increment()
counter.increment()
counter.increment()
counter.decrement()
counter.get_count()
counter.get_count()
counter.reset()
counter.increment()

# Check usage statistics
usage_stats = counter.get_usage_stats()
print("Method usage statistics:")
for method, count in usage_stats.items():
    print(f"  {method}: {count} calls")
````

## Troubleshooting / FAQs

### Q: What's the difference between methods and attributes?

A: In Python, methods are callable attributes of an object. The functions in the `methodnames` module specifically return names of attributes that are callable (i.e., methods), not all attributes.

### Q: Why use the iterator versions of the functions?

A: The iterator versions (`iter_method_names` and `iter_public_method_names`) are more memory-efficient for objects with many methods, as they generate method names on-demand rather than creating a complete list upfront. This can be important when working with large objects or in memory-constrained environments.

### Q: What's considered a "public" method?

A: In Python, there's no strict enforcement of public/private methods, but by convention, methods whose names start with an underscore (`_`) are considered private or internal. The `iter_public_method_names` and `get_public_method_names` functions follow this convention and return only methods whose names don't start with an underscore.

### Q: Will these functions return inherited methods?

A: Yes, the functions will return all methods available on the object, including those inherited from parent classes. This is because they use Python's `dir()` function internally, which includes inherited attributes.
