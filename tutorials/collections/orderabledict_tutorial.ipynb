# Using the OrderableDict Module in baseobjects

## Introduction

The `OrderableDict` module provides a dictionary-like container that combines the functionality of a dictionary with the ability to perform operations based on the order of items. It extends the `BaseDict` class from the baseobjects package, adding methods for inserting, moving, and accessing items by their position in the dictionary.

Unlike Python's built-in `OrderedDict` which simply remembers the insertion order, `OrderableDict` provides methods to explicitly manipulate the order of items, such as inserting items at specific positions, moving existing items, and accessing items by their index.

This tutorial will guide you through:
- Understanding the purpose and design of `OrderableDict`
- Using the dictionary-like functionality
- Manipulating the order of items
- Advanced features and use cases

**Prerequisites:**
- Basic understanding of Python dictionaries
- Familiarity with Python's `OrderedDict` from the collections module
- Knowledge of the `BaseDict` class from the baseobjects package

## Importing the Module

```
from baseobjects.collections.orderabledict import OrderableDict
import copy
```

## Core Functionality

The `OrderableDict` class provides a dictionary-like container with methods for manipulating the order of items. Let's explore its core functionality.

### Key Features

1. **Dictionary-Like Behavior**: Provides all standard dictionary operations
2. **Index-Based Access**: Access items by their position in the dictionary
3. **Order Manipulation**: Insert, move, and reorder items
4. **Inheritance from BaseDict**: Provides proper copying and deep copying support

Let's create a simple `OrderableDict` and explore its functionality:

```
# Create an empty OrderableDict
orderable_dict = OrderableDict()
print(f"Empty OrderableDict: {orderable_dict}")

# Add some items
orderable_dict["a"] = 1
orderable_dict["b"] = 2
orderable_dict["c"] = 3
orderable_dict["d"] = 4
print(f"OrderableDict with items: {orderable_dict}")

# Access items by key
print(f"Value of 'a': {orderable_dict['a']}")
print(f"Value of 'c': {orderable_dict['c']}")

# Check if a key exists
print(f"'b' in orderable_dict: {'b' in orderable_dict}")
print(f"'e' in orderable_dict: {'e' in orderable_dict}")

# Get the number of items
print(f"len(orderable_dict): {len(orderable_dict)}")
```

### Accessing Items by Index

One of the key features of `OrderableDict` is the ability to access items by their position in the dictionary:

```
# Access items by index
print(f"First item (index 0): {orderable_dict.get_index(0)}")
print(f"Second item (index 1): {orderable_dict.get_index(1)}")
print(f"Last item (index -1): {orderable_dict.get_index(-1)}")

# Get a default value if the index is out of range
print(f"Index 10 (out of range): {orderable_dict.get_index(10, 'default')}")

# Iterate through the dictionary
print("\nIterating through the dictionary:")
for key in orderable_dict:
    print(f"{key}: {orderable_dict[key]}")
```

### Modifying Items by Index

You can modify items by their position in the dictionary:

```
# Set the value of an item by index
orderable_dict.set_index(1, 22)  # Change the value of the second item
print(f"After setting index 1: {orderable_dict}")

# Insert a new item at a specific position
orderable_dict.insert(2, "x", 99)  # Insert ('x', 99) at index 2
print(f"After inserting at index 2: {orderable_dict}")

# Insert and move an existing key to a new position
orderable_dict.insert_move(0, "c", 33)  # Move 'c' to index 0 and update its value
print(f"After moving 'c' to index 0: {orderable_dict}")
```

### Dictionary Methods

`OrderableDict` inherits all the standard dictionary methods from `BaseDict`:

```
# Get all keys
print(f"orderable_dict.keys(): {orderable_dict.keys()}")

# Get all values
print(f"orderable_dict.values(): {orderable_dict.values()}")

# Get all items
print(f"orderable_dict.items(): {orderable_dict.items()}")

# Get a value with a default if the key doesn't exist
print(f"orderable_dict.get('a'): {orderable_dict.get('a')}")
print(f"orderable_dict.get('z', 'default'): {orderable_dict.get('z', 'default')}")

# Set a default value if the key doesn't exist
print(f"orderable_dict.setdefault('e', 5): {orderable_dict.setdefault('e', 5)}")
print(f"After setdefault: {orderable_dict}")
```

### Removing Items

`OrderableDict` provides methods for removing items:

```
# Pop an item by key
popped_value = orderable_dict.pop("a")
print(f"Popped value: {popped_value}")
print(f"After pop('a'): {orderable_dict}")

# Pop an item by index
popped_value = orderable_dict.pop_index(1)
print(f"Popped item at index 1: {popped_value}")
print(f"After pop_index(1): {orderable_dict}")

# Pop the last item
popped_item = orderable_dict.popitem()
print(f"Popped last item: {popped_item}")
print(f"After popitem(): {orderable_dict}")

# Remove an item by key
orderable_dict["f"] = 6  # Add a new item
print(f"After adding 'f': {orderable_dict}")
orderable_dict.remove("f")
print(f"After remove('f'): {orderable_dict}")

# Clear all items
orderable_dict_copy = copy.copy(orderable_dict)
orderable_dict_copy.clear()
print(f"After clear(): {orderable_dict_copy}")
print(f"Original: {orderable_dict}")
```

## Module Interaction

The `OrderableDict` module interacts with other modules in the baseobjects package, particularly `BaseDict`. These interactions provide enhanced functionality for the orderable dictionary.

### Interaction with BaseDict

`OrderableDict` inherits from `BaseDict`, which means it also inherits all the functionality of `BaseDict`, including copying and deep copying:

```
# Create a new OrderableDict
order_dict = OrderableDict()
order_dict["a"] = 1
order_dict["b"] = 2
order_dict["c"] = 3

# Create a copy of the OrderableDict
order_dict_copy = order_dict.copy()
print(f"Original: {order_dict}")
print(f"Copy: {order_dict_copy}")
print(f"Are they the same object? {order_dict is order_dict_copy}")

# Modify the copy
order_dict_copy["d"] = 4
print(f"\nAfter modifying the copy:")
print(f"Original: {order_dict}")
print(f"Copy: {order_dict_copy}")
```

### Deep Copying

`OrderableDict` also supports deep copying through inheritance from `BaseDict`:

```
# Create an OrderableDict with nested structures
nested_dict = OrderableDict()
nested_dict["simple"] = 1
nested_dict["list"] = [1, 2, 3]
nested_dict["dict"] = {"a": 1, "b": 2}
print(f"Nested dict: {nested_dict}")

# Create a deep copy
nested_dict_deepcopy = copy.deepcopy(nested_dict)
print(f"Deep copy: {nested_dict_deepcopy}")
print(f"Are they the same object? {nested_dict is nested_dict_deepcopy}")
print(f"Is the nested list the same object? {nested_dict['list'] is nested_dict_deepcopy['list']}")
print(f"Is the nested dict the same object? {nested_dict['dict'] is nested_dict_deepcopy['dict']}")

# Modify the nested structures in the deep copy
nested_dict_deepcopy["list"].append(4)
nested_dict_deepcopy["dict"]["c"] = 3

print(f"\nAfter modifying the deep copy:")
print(f"Original list: {nested_dict['list']}")
print(f"Original dict: {nested_dict['dict']}")
print(f"Deep copy list: {nested_dict_deepcopy['list']}")
print(f"Deep copy dict: {nested_dict_deepcopy['dict']}")
```

## Advanced Features

The `OrderableDict` class provides several advanced features that make it powerful for specific use cases.

### Appending Items

You can append items to the end of the dictionary:

```
# Create a new OrderableDict
append_dict = OrderableDict()
append_dict["a"] = 1
append_dict["b"] = 2
print(f"Initial dictionary: {append_dict}")

# Append a new item
append_dict.append("c", 3)
print(f"After appending ('c', 3): {append_dict}")

# Append multiple items using update
append_dict.update({"d": 4, "e": 5})
print(f"After updating with multiple items: {append_dict}")
```

### Reversing the Order

You can reverse the order of items in the dictionary:

```
# Create a new OrderableDict
reverse_dict = OrderableDict()
reverse_dict["a"] = 1
reverse_dict["b"] = 2
reverse_dict["c"] = 3
reverse_dict["d"] = 4
print(f"Initial dictionary: {reverse_dict}")

# Reverse the order
reverse_dict.reverse()
print(f"After reversing: {reverse_dict}")
```

### Combining OrderableDict with Other Dictionary Types

`OrderableDict` can be initialized with other dictionary types:

```
# Create a dictionary
regular_dict = {"a": 1, "b": 2, "c": 3}

# Create an OrderableDict from a regular dictionary
from_dict = OrderableDict(regular_dict)
print(f"OrderableDict from regular dict: {from_dict}")

# Create an OrderableDict from keyword arguments
from_kwargs = OrderableDict(d=4, e=5, f=6)
print(f"OrderableDict from kwargs: {from_kwargs}")

# Create an OrderableDict from a combination of sources
combined = OrderableDict(regular_dict, g=7, h=8)
print(f"Combined OrderableDict: {combined}")
```

### Implementing a Simple Configuration Manager

`OrderableDict` is useful for implementing a configuration manager where the order of settings matters:

```
class ConfigManager:
    """A simple configuration manager with ordered settings."""
    
    def __init__(self):
        self.config = OrderableDict()
        
        # Add default settings in a specific order
        self.config["debug"] = False
        self.config["log_level"] = "INFO"
        self.config["max_connections"] = 10
        self.config["timeout"] = 30
        self.config["retry_attempts"] = 3
    
    def get_setting(self, key, default=None):
        """Get a configuration setting."""
        return self.config.get(key, default)
    
    def set_setting(self, key, value):
        """Set a configuration setting."""
        self.config[key] = value
    
    def add_setting_at_beginning(self, key, value):
        """Add a setting at the beginning of the configuration."""
        self.config.insert_move(0, key, value)
    
    def add_setting_before(self, before_key, key, value):
        """Add a setting before another setting."""
        try:
            # Find the index of the before_key
            keys = list(self.config.keys())
            index = keys.index(before_key)
            self.config.insert(index, key, value)
        except ValueError:
            # If before_key doesn't exist, append to the end
            self.config[key] = value
    
    def add_setting_after(self, after_key, key, value):
        """Add a setting after another setting."""
        try:
            # Find the index of the after_key
            keys = list(self.config.keys())
            index = keys.index(after_key)
            self.config.insert(index + 1, key, value)
        except ValueError:
            # If after_key doesn't exist, append to the end
            self.config[key] = value
    
    def remove_setting(self, key):
        """Remove a configuration setting."""
        if key in self.config:
            del self.config[key]
    
    def get_all_settings(self):
        """Get all configuration settings in order."""
        return self.config
    
    def __str__(self):
        result = "Configuration Settings:\n"
        for key, value in self.config.items():
            result += f"  {key}: {value}\n"
        return result

# Create a configuration manager
config_manager = ConfigManager()
print(config_manager)

# Modify some settings
config_manager.set_setting("log_level", "DEBUG")
config_manager.set_setting("max_connections", 20)
print("\nAfter modifying settings:")
print(config_manager)

# Add settings at specific positions
config_manager.add_setting_at_beginning("app_name", "MyApp")
print("\nAfter adding setting at beginning:")
print(config_manager)

config_manager.add_setting_before("timeout", "connection_type", "TCP")
print("\nAfter adding setting before 'timeout':")
print(config_manager)

config_manager.add_setting_after("timeout", "keep_alive", True)
print("\nAfter adding setting after 'timeout':")
print(config_manager)

# Remove a setting
config_manager.remove_setting("retry_attempts")
print("\nAfter removing 'retry_attempts':")
print(config_manager)
```

### Implementing a Form Builder

Another practical use case for `OrderableDict` is implementing a form builder where the order of fields matters:

```
class FormField:
    """A field in a form."""
    
    def __init__(self, name, field_type, label=None, required=False, default=None, options=None):
        self.name = name
        self.field_type = field_type
        self.label = label or name.replace("_", " ").title()
        self.required = required
        self.default = default
        self.options = options
    
    def __str__(self):
        options_str = f", options={self.options}" if self.options else ""
        return f"Field(name='{self.name}', type='{self.field_type}', label='{self.label}', required={self.required}, default={self.default}{options_str})"

class FormBuilder:
    """A form builder with ordered fields."""
    
    def __init__(self, form_name):
        self.form_name = form_name
        self.fields = OrderableDict()
    
    def add_field(self, name, field_type, label=None, required=False, default=None, options=None):
        """Add a field to the form."""
        field = FormField(name, field_type, label, required, default, options)
        self.fields[name] = field
        return field
    
    def add_field_at_position(self, position, name, field_type, label=None, required=False, default=None, options=None):
        """Add a field at a specific position in the form."""
        field = FormField(name, field_type, label, required, default, options)
        self.fields.insert(position, name, field)
        return field
    
    def move_field(self, name, new_position):
        """Move a field to a new position in the form."""
        if name in self.fields:
            field = self.fields[name]
            self.fields.insert_move(new_position, name, field)
    
    def remove_field(self, name):
        """Remove a field from the form."""
        if name in self.fields:
            del self.fields[name]
    
    def get_field(self, name):
        """Get a field by name."""
        return self.fields.get(name)
    
    def get_field_at_position(self, position):
        """Get a field at a specific position."""
        return self.fields.get_index(position)
    
    def get_all_fields(self):
        """Get all fields in order."""
        return self.fields
    
    def __str__(self):
        result = f"Form: {self.form_name}\n"
        for i, (name, field) in enumerate(self.fields.items()):
            result += f"  {i}. {field}\n"
        return result

# Create a form builder
form_builder = FormBuilder("User Registration")

# Add fields
form_builder.add_field("username", "text", required=True)
form_builder.add_field("email", "email", required=True)
form_builder.add_field("password", "password", required=True)
form_builder.add_field("confirm_password", "password", required=True)
form_builder.add_field("age", "number", required=False)
form_builder.add_field("gender", "select", required=False, options=["Male", "Female", "Other"])

print(form_builder)

# Add a field at a specific position
form_builder.add_field_at_position(0, "title", "select", required=True, options=["Mr.", "Mrs.", "Ms.", "Dr."])
print("\nAfter adding 'title' field at position 0:")
print(form_builder)

# Move a field to a new position
form_builder.move_field("gender", 2)
print("\nAfter moving 'gender' field to position 2:")
print(form_builder)

# Remove a field
form_builder.remove_field("age")
print("\nAfter removing 'age' field:")
print(form_builder)
```

## API Highlights

Here are the key components of the `OrderableDict` module API:

### OrderableDict
- `__init__(dict_=None, /, *args, **kwargs)`: Initialize a new OrderableDict instance
- `get_index(index, default=SENTINEL)`: Get the (key, value) pair at the specified index
- `set_index(index, value)`: Set the value of the item at the specified index
- `insert(index, key, value)`: Insert a new (key, value) pair at the specified index
- `insert_move(index, key, value)`: Move an existing key to the specified index and update its value
- `append(key, value)`: Append a new (key, value) pair to the end of the dictionary
- `pop(key)`: Remove and return the value for the specified key
- `pop_index(index=-1)`: Remove and return the (key, value) pair at the specified index
- `popitem()`: Remove and return the last (key, value) pair
- `remove(key)`: Remove the specified key
- `clear()`: Remove all items
- `reverse()`: Reverse the order of items
- All standard dictionary methods inherited from `BaseDict`

For more detailed information, consult the full API documentation.

## Troubleshooting / FAQs

### Q: When should I use OrderableDict instead of OrderedDict?

A: `OrderableDict` is useful when you need to explicitly manipulate the order of items in a dictionary, such as inserting items at specific positions, moving existing items, or accessing items by their index. If you only need to maintain insertion order without these additional capabilities, Python's built-in `OrderedDict` might be sufficient.

### Q: How does OrderableDict handle copying and deep copying?

A: `OrderableDict` inherits from `BaseDict`, which properly implements Python's copy protocol through `__copy__` and `__deepcopy__` methods. This ensures that both the dictionary and its order are properly copied.

### Q: Can I use negative indices with OrderableDict?

A: Yes, negative indices work similar to lists, with -1 referring to the last item, -2 to the second-to-last item, and so on.

### Q: What happens if I try to access an index that is out of range?

A: If you try to access an index that is out of range using `get_index()`, it will return the default value if provided, or raise a `KeyError` if no default is provided. If you try to set an index that is out of range using `set_index()`, it will raise an `IndexError`.
