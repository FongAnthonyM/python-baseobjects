# Using the DeepChainMap Module in baseobjects

## Introduction

The `DeepChainMap` module provides a specialized version of Python's standard `ChainMap` that offers more intuitive update and delete behavior. Unlike the standard `ChainMap`, which always updates or adds items to the first mapping, `DeepChainMap` updates items in the first mapping that contains the key, and only adds new items to the first mapping.

This behavior is particularly useful for configuration scenarios where you want to modify existing values in their original location rather than always modifying the first mapping. Note that DeepChainMap only works with top-level keys and cannot handle nested dictionary management.

This tutorial will guide you through:
- Understanding the purpose and design of `DeepChainMap`
- Using the basic functionality of `DeepChainMap`
- Comparing `DeepChainMap` with the standard `ChainMap`
- Exploring advanced features and practical use cases

**Prerequisites:**
- Basic understanding of Python dictionaries
- Familiarity with Python's `ChainMap` from the collections module
- Knowledge of the `BaseObject` class from the baseobjects package

## Importing the Module

```
# Import the standard ChainMap for comparison
from collections import ChainMap

# Import DeepChainMap and BaseObject from baseobjects
from baseobjects.collections import DeepChainMap
from baseobjects.bases import BaseObject
```

## Core Functionality

The `DeepChainMap` class provides a more intuitive way to work with a chain of mappings. Let's explore its core functionality.

### Key Features

1. **Chain of Mappings**: Like `ChainMap`, it maintains a list of mappings that are searched in order.
2. **Intuitive Updates**: Updates existing keys in the first mapping that contains the key.
3. **Intuitive Deletions**: Deletes keys from the first mapping that contains the key.
4. **New Keys**: Adds new keys to the first mapping.
5. **Top-Level Keys Only**: Works only with top-level keys and does not provide special handling for nested dictionaries.

Let's create a simple `DeepChainMap` and explore its functionality:

```
# Create dictionaries for the chain
defaults = {"theme": "default", "language": "en", "timeout": 30, "debug": False}
user_settings = {"language": "fr", "timeout": 60}
session_settings = {"theme": "dark"}

# Create a DeepChainMap with these dictionaries
settings = DeepChainMap(session_settings, user_settings, defaults)

# Access values (looks through the chain)
print("Accessing values:")
print(f"  Theme: {settings['theme']}")
print(f"  Language: {settings['language']}")
print(f"  Timeout: {settings['timeout']}")
print(f"  Debug: {settings['debug']}")

# Check the original dictionaries
print("\nOriginal dictionaries:")
print(f"  Session settings: {session_settings}")
print(f"  User settings: {user_settings}")
print(f"  Defaults: {defaults}")
```

### Adding and Updating Items

One of the key features of `DeepChainMap` is how it handles adding and updating items:

```
# Add a new key
print("Adding a new key 'notifications':")
settings["notifications"] = True
print(f"  Settings['notifications']: {settings['notifications']}")
print(f"  First mapping (session_settings): {session_settings}")

# Update an existing key
print("\nUpdating existing key 'language':")
settings["language"] = "es"
print(f"  Settings['language']: {settings['language']}")
print(f"  User settings: {user_settings}")
print(f"  Defaults (unchanged): {defaults}")
```

### Deleting Items

`DeepChainMap` also provides intuitive deletion behavior:

```
# Delete a key
print("Deleting key 'timeout':")
del settings["timeout"]
print(f"  'timeout' in settings: {'timeout' in settings}")
print(f"  User settings: {user_settings}")

# Try to access a deleted key that exists in defaults
print("\nAccessing 'timeout' after deletion from user_settings:")
print(f"  Settings['timeout']: {settings['timeout']}")
```

## Module Interaction

The `DeepChainMap` module interacts with other modules in the baseobjects package, particularly `BaseObject`. These interactions provide enhanced functionality for the deep chain map.

### Interaction with BaseObject

`DeepChainMap` inherits from `BaseObject`, which provides additional functionality:

```
# Create a new DeepChainMap
map1 = {"a": 1, "b": 2}
map2 = {"c": 3, "d": 4}
deep_chain = DeepChainMap(map1, map2)

# Check if it's an instance of BaseObject
print(f"Is deep_chain an instance of BaseObject? {isinstance(deep_chain, BaseObject)}")

# Use BaseObject methods
print(f"Class name: {deep_chain.__class__.__name__}")
```

### Interaction with ChainMap

`DeepChainMap` also inherits from `ChainMap`, which means it has all the functionality of `ChainMap`:

```
# Access the maps attribute
print(f"Maps: {deep_chain.maps}")

# Use ChainMap methods
print(f"Keys: {list(deep_chain.keys())}")
print(f"Values: {list(deep_chain.values())}")
print(f"Items: {list(deep_chain.items())}")

# Create a new child context
child = deep_chain.new_child({"e": 5})
print(f"Child maps: {child.maps}")
print(f"Child['e']: {child['e']}")
```

## Advanced Features

The `DeepChainMap` class provides several advanced features that make it powerful for specific use cases.

### Comparison with Standard ChainMap

Let's compare `DeepChainMap` with the standard `ChainMap` to understand the key differences:

```
# Create dictionaries for the chains
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 30}
dict3 = {"c": 300, "d": 400}

# Create a standard ChainMap
standard_chain = ChainMap(dict1, dict2, dict3)

# Create a DeepChainMap with the same dictionaries
deep_chain = DeepChainMap(dict1, dict2, dict3)

print("Initial dictionaries:")
print(f"  dict1: {dict1}")
print(f"  dict2: {dict2}")
print(f"  dict3: {dict3}")

print("\nAccessing values (both work the same):")
print(f"  standard_chain['a']: {standard_chain['a']}")
print(f"  deep_chain['a']: {deep_chain['a']}")
print(f"  standard_chain['b']: {standard_chain['b']}")
print(f"  deep_chain['b']: {deep_chain['b']}")
print(f"  standard_chain['c']: {standard_chain['c']}")
print(f"  deep_chain['c']: {deep_chain['c']}")
```

### Difference in Update Behavior

The key difference between `DeepChainMap` and `ChainMap` is in how they handle updates:

```
# Update existing key 'b'
standard_chain["b"] = 200
deep_chain["b"] = 200

print("After update:")
print(f"  standard_chain['b']: {standard_chain['b']}")
print(f"  deep_chain['b']: {deep_chain['b']}")

print("\nDifference in behavior - which dictionary was updated:")
print(f"  dict1 with standard_chain: {dict1}")
print(f"  dict1 with deep_chain: {dict1}")
print(f"  standard_chain.maps[0]: {standard_chain.maps[0]}")

# Reset dict1 for the next test
dict1["b"] = 2

# Update key 'c' (exists in dict2 and dict3)
standard_chain["c"] = 3000
deep_chain["c"] = 3000

print("\nUpdating key 'c' (exists in dict2 and dict3):")
print(f"  dict1 with standard_chain: {dict1}")
print(f"  dict2 with standard_chain: {dict2}")
print(f"  dict1 with deep_chain: {dict1}")
print(f"  dict2 with deep_chain: {dict2}")
```

### Difference in Delete Behavior

There's also a difference in how `DeepChainMap` and `ChainMap` handle deletions:

```
# Reset for deletion test
dict2["c"] = 30

print("Deleting key 'c':")
try:
    del standard_chain["c"]
    print("  Deleted 'c' from standard_chain")
except KeyError:
    print("  KeyError: Cannot delete 'c' from standard_chain")

del deep_chain["c"]
print("  Deleted 'c' from deep_chain")

print("After deletion:")
print(f"  dict2 with standard_chain: {dict2}")
print(f"  dict2 with deep_chain: {dict2}")
print(f"  'c' in standard_chain: {'c' in standard_chain}")
print(f"  'c' in deep_chain: {'c' in deep_chain}")
print(f"  deep_chain['c']: {deep_chain['c']}")
```

## Examples

Let's explore some practical examples of using the `DeepChainMap` module.

### Configuration Management Example

`DeepChainMap` is particularly useful for configuration management with multiple layers of settings. In this example, we'll use flat dictionaries with compound keys (like "db_host" instead of nested structures) since DeepChainMap only works with top-level keys:

```
# System defaults
system_defaults = {
    "db_host": "localhost",
    "db_port": 5432,
    "db_name": "app_db",
    "db_user": "app_user",
    "db_password": "default_password",
    "api_url": "https://api.example.com",
    "api_timeout": 30,
    "api_retry_attempts": 3,
    "log_level": "INFO",
    "log_file": "app.log",
}

# User configuration (overrides some defaults)
user_config = {
    "db_host": "db.example.com",
    "db_password": "user_password",
    "log_level": "DEBUG",
}

# Environment-specific overrides
env_overrides = {
    "api_url": "https://staging-api.example.com",
}

# Create configuration using DeepChainMap
config = DeepChainMap(env_overrides, user_config, system_defaults)

print("Effective configuration:")
print("Database settings:")
print(f"  Host: {config['db_host']}")
print(f"  Port: {config['db_port']}")
print(f"  Name: {config['db_name']}")
print(f"  User: {config['db_user']}")
print(f"  Password: {config['db_password']}")

print("\nAPI settings:")
print(f"  URL: {config['api_url']}")
print(f"  Timeout: {config['api_timeout']}")
print(f"  Retry attempts: {config['api_retry_attempts']}")

print("\nLogging settings:")
print(f"  Level: {config['log_level']}")
print(f"  File: {config['log_file']}")
```

### Updating Configuration Settings

You can update configuration settings, and the changes will be applied to the appropriate mapping:

```
# Update a setting
print("Updating API timeout:")
config["api_timeout"] = 60

print("After update:")
print(f"  API timeout: {config['api_timeout']}")
print(f"  env_overrides: {env_overrides}")
```

### Feature Flag Management Example

Another practical use case for `DeepChainMap` is managing feature flags with different levels of override:

```
# Default feature flags
default_flags = {
    "new_ui": False,
    "advanced_search": False,
    "analytics": True,
    "notifications": True,
}

# User-specific feature flags
user_flags = {
    "new_ui": True,
    "analytics": False,
}

# A/B testing overrides
ab_test_flags = {
    "advanced_search": True,
}

# Create feature flags using DeepChainMap
feature_flags = DeepChainMap(ab_test_flags, user_flags, default_flags)

print("Feature flags:")
print(f"  New UI: {feature_flags['new_ui']}")
print(f"  Advanced Search: {feature_flags['advanced_search']}")
print(f"  Analytics: {feature_flags['analytics']}")
print(f"  Notifications: {feature_flags['notifications']}")

# Update a feature flag
print("\nUpdating 'notifications' flag:")
feature_flags["notifications"] = False
print(f"  Notifications: {feature_flags['notifications']}")
print(f"  ab_test_flags: {ab_test_flags}")
```

## API Highlights

Here are the key components of the `DeepChainMap` module API:

### DeepChainMap
- `__init__(*args, **kwargs)`: Initialize a new DeepChainMap instance
- `__setitem__(key, value)`: Set an item, updating the first mapping that contains the key or adding to the first mapping if the key doesn't exist
- `__delitem__(key)`: Delete an item from the first mapping that contains it
- All methods inherited from `ChainMap`, including:
  - `maps`: A list of mappings that make up the chain
  - `new_child(m=None)`: Create a new child context with a new map
  - `parents`: The parents of this mapping (all mappings except the first)
  - `get(key, default=None)`: Get a value with a default if the key doesn't exist
  - `keys()`, `values()`, `items()`: Standard dictionary methods

For more detailed information, consult the full API documentation.

## Troubleshooting / FAQs

### Q: When should I use DeepChainMap instead of ChainMap?

A: Use `DeepChainMap` when you want updates and deletions to affect the first mapping that contains the key, rather than always affecting the first mapping. This is particularly useful for configuration management where you want to modify existing values in their original location.

### Q: How does DeepChainMap handle nested dictionaries?

A: `DeepChainMap` does not support nested dictionary management. It only works with top-level keys in the mappings. If a value in one of the mappings is itself a dictionary, DeepChainMap will treat it as a regular value and won't provide any special handling for its nested keys. To work with nested structures, it's recommended to use flat dictionaries with compound keys (like "db_host" instead of "database": {"host": ...}) or use alternative approaches for complex nested configurations.

### Q: Can I use DeepChainMap with non-dictionary mappings?

A: Yes, `DeepChainMap` works with any mapping type, not just dictionaries. However, the mappings must support the standard mapping interface, including `__getitem__`, `__setitem__`, and `__delitem__`.

### Q: What happens if I try to delete a key that doesn't exist?

A: If you try to delete a key that doesn't exist in any mapping, `DeepChainMap` will raise a `KeyError`, just like a regular dictionary.
