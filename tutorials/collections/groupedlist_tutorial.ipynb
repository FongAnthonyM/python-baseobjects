Groups can have hierarchical relationships, with parent-child connections between groups.

## Importing the Module

````python
from baseobjects.collections.groupedlist import GroupedList
import copy
````

## Core Functionality

The `GroupedList` class provides a list-like container that organizes items into named groups. Let's explore its core functionality.

### Key Features

1. **Named Groups**: Organize items into named groups within a single list
2. **Hierarchical Structure**: Create parent-child relationships between groups
3. **Group-Specific Operations**: Perform operations on specific groups
4. **List-Like Interface**: Use standard list operations on the entire list or specific groups

Let's create a simple `GroupedList` and explore its functionality:

````python
# Create an empty GroupedList
grouped_list = GroupedList()
print(f"Empty GroupedList: {grouped_list}")

# Add some items to the main list
grouped_list.append("Item 1")
grouped_list.append("Item 2")
grouped_list.append("Item 3")
print(f"GroupedList with items: {grouped_list}")
print(f"Length: {len(grouped_list)}")
````

### Creating and Managing Groups

You can create named groups within the `GroupedList` and add items to them:

````python
# Create a group named "fruits"
fruits_group = grouped_list.create_group("fruits")
fruits_group.append("Apple")
fruits_group.append("Banana")
fruits_group.append("Orange")

# Create a group named "vegetables"
vegetables_group = grouped_list.create_group("vegetables")
vegetables_group.append("Carrot")
vegetables_group.append("Broccoli")
vegetables_group.append("Spinach")

print(f"Main list: {grouped_list}")
print(f"Fruits group: {fruits_group}")
print(f"Vegetables group: {vegetables_group}")

# Get the length of each group
print(f"Main list length: {len(grouped_list)}")
print(f"Fruits group length: {len(fruits_group)}")
print(f"Vegetables group length: {len(vegetables_group)}")

# Get the total length of all groups
group_lengths = grouped_list.get_group_lengths()
print(f"Group lengths: {group_lengths}")
print(f"Total items (including groups): {grouped_list.get_length()}")
````

### Accessing Groups and Items

You can access groups and items within groups using various methods:

````python
# Get a group by name
fruits = grouped_list.get_group("fruits")
print(f"Retrieved fruits group: {fruits}")

# Access items in a group
print(f"First fruit: {fruits[0]}")
print(f"Second fruit: {fruits[1]}")
print(f"Last fruit: {fruits[-1]}")

# Access items in the main list
print(f"First item in main list: {grouped_list[0]}")
print(f"Second item in main list: {grouped_list[1]}")

# Check if an item is in a group
print(f"'Apple' in fruits: {'Apple' in fruits}")
print(f"'Carrot' in fruits: {'Carrot' in fruits}")

# Iterate through a group
print("\nIterating through fruits group:")
for fruit in fruits:
    print(fruit)
````

### Modifying Groups and Items

You can modify groups and items within groups using various methods:

````python
# Add an item to a group
fruits.append("Grape")
print(f"Fruits after adding 'Grape': {fruits}")

# Insert an item at a specific position in a group
fruits.insert(1, "Pineapple")
print(f"Fruits after inserting 'Pineapple' at position 1: {fruits}")

# Remove an item from a group
fruits.remove("Banana")
print(f"Fruits after removing 'Banana': {fruits}")

# Pop an item from a group
popped_fruit = fruits.pop()
print(f"Popped fruit: {popped_fruit}")
print(f"Fruits after popping: {fruits}")

# Clear a group
vegetables_copy = copy.copy(vegetables_group)
vegetables_copy.clear()
print(f"Vegetables copy after clearing: {vegetables_copy}")
print(f"Original vegetables group: {vegetables_group}")
````

### Group-Specific Operations

You can perform operations on specific groups:

````python
# Create a new group for dairy products
dairy_group = grouped_list.create_group("dairy")
dairy_group.append("Milk")
dairy_group.append("Cheese")
dairy_group.append("Yogurt")

# Extend a group with multiple items
dairy_group.extend(["Butter", "Cream"])
print(f"Dairy group after extending: {dairy_group}")

# Get an item from a specific group using the main list
apple = grouped_list.get_item(0, group="fruits")
print(f"First item in fruits group: {apple}")

# Set an item in a specific group using the main list
grouped_list.set_item(0, "Green Apple", group="fruits")
print(f"Fruits group after setting first item: {fruits}")

# Delete an item from a specific group using the main list
grouped_list.delete_item(0, group="dairy")
print(f"Dairy group after deleting first item: {dairy_group}")
````

## Module Interaction

The `GroupedList` module interacts with other modules in the baseobjects package, particularly `BaseList`. These interactions provide enhanced functionality for the grouped list.

### Interaction with BaseList

`GroupedList` inherits from `BaseList`, which means it also inherits all the functionality of `BaseList`, including copying and standard list operations:

````python
# Create a copy of the GroupedList
grouped_list_copy = grouped_list.copy()
print(f"Original: {grouped_list}")
print(f"Copy: {grouped_list_copy}")
print(f"Are they the same object? {grouped_list is grouped_list_copy}")

# Modify the copy
grouped_list_copy.append("New Item")
print("\nAfter modifying the copy:")
print(f"Original: {grouped_list}")
print(f"Copy: {grouped_list_copy}")

# Check if the groups are copied
fruits_copy = grouped_list_copy.get_group("fruits")
print(f"\nFruits group in copy: {fruits_copy}")
print(f"Are the fruit groups the same object? {fruits is fruits_copy}")

# Modify the fruits group in the copy
fruits_copy.append("Mango")
print("\nAfter modifying the fruits group in the copy:")
print(f"Original fruits group: {fruits}")
print(f"Copy fruits group: {fruits_copy}")
````

### Deep Copying

`GroupedList` also supports deep copying through inheritance from `BaseList`:

````python
# Create a GroupedList with nested lists
nested_grouped_list = GroupedList()
nested_grouped_list.append([1, 2, 3])
nested_grouped_list.append({"key": "value"})

# Create a nested group
nested_group = nested_grouped_list.create_group("nested")
nested_group.append([4, 5, 6])
nested_group.append({"nested_key": "nested_value"})

print(f"Nested GroupedList: {nested_grouped_list}")
print(f"Nested group: {nested_group}")

# Create a deep copy
nested_grouped_list_deepcopy = copy.deepcopy(nested_grouped_list)
print(f"Deep copy: {nested_grouped_list_deepcopy}")
print(f"Are they the same object? {nested_grouped_list is nested_grouped_list_deepcopy}")

# Modify the nested list in the deep copy
nested_grouped_list_deepcopy[0].append(4)
nested_grouped_list_deepcopy[1]["new_key"] = "new_value"

# Get the nested group in the deep copy
nested_group_deepcopy = nested_grouped_list_deepcopy.get_group("nested")
nested_group_deepcopy[0].append(7)
nested_group_deepcopy[1]["new_nested_key"] = "new_nested_value"

print("\nAfter modifying the deep copy:")
print(f"Original first element: {nested_grouped_list[0]}")
print(f"Original second element: {nested_grouped_list[1]}")
print(f"Deep copy first element: {nested_grouped_list_deepcopy[0]}")
print(f"Deep copy second element: {nested_grouped_list_deepcopy[1]}")

print(f"\nOriginal nested group first element: {nested_group[0]}")
print(f"Original nested group second element: {nested_group[1]}")
print(f"Deep copy nested group first element: {nested_group_deepcopy[0]}")
print(f"Deep copy nested group second element: {nested_group_deepcopy[1]}")
````

## Advanced Features

The `GroupedList` class provides several advanced features that make it powerful for specific use cases.

### Hierarchical Group Relationships

You can create hierarchical relationships between groups, with parent-child connections:

````python
# Create a new GroupedList for a hierarchical example
hierarchy_list = GroupedList()

# Create parent groups
animals = hierarchy_list.create_group("animals")
plants = hierarchy_list.create_group("plants")

# Create child groups for animals
mammals = animals.create_group("mammals")
birds = animals.create_group("birds")
fish = animals.create_group("fish")

# Create child groups for plants
trees = plants.create_group("trees")
flowers = plants.create_group("flowers")

# Add items to the groups
mammals.extend(["Dog", "Cat", "Elephant"])
birds.extend(["Eagle", "Sparrow", "Penguin"])
fish.extend(["Salmon", "Tuna", "Shark"])
trees.extend(["Oak", "Pine", "Maple"])
flowers.extend(["Rose", "Tulip", "Daisy"])

# Check parent-child relationships
print(f"Is 'mammals' a child of 'animals'? {animals.check_if_child(mammals)}")
print(f"Is 'animals' a parent of 'mammals'? {mammals.check_if_parent(animals)}")
print(f"Is 'trees' a child of 'animals'? {animals.check_if_child(trees)}")

# Get group lengths with recursion (including child groups)
group_lengths_recursive = hierarchy_list.get_group_lengths(recurse=True)
print(f"\nGroup lengths (recursive): {group_lengths_recursive}")
print(f"Total items (including all groups and subgroups): {hierarchy_list.get_length()}")

# Access items through the hierarchy
print(f"\nFirst mammal: {mammals[0]}")
print(f"Second bird: {birds[1]}")
print(f"Third fish: {fish[2]}")
````

### Multiple Parents

A group can have multiple parents:

````python
# Create a new GroupedList for a multiple parents example
multi_parent_list = GroupedList()

# Create parent groups
carnivores = multi_parent_list.create_group("carnivores")
aquatic = multi_parent_list.create_group("aquatic")
domestic = multi_parent_list.create_group("domestic")

# Create a group with multiple parents
pets = carnivores.create_group("pets", parents=[domestic])
aquatic_carnivores = carnivores.create_group("aquatic_carnivores", parents=[aquatic])

# Add items to the groups
pets.extend(["Dog", "Cat", "Ferret"])
aquatic_carnivores.extend(["Shark", "Piranha", "Killer Whale"])

# Check parent-child relationships
print(f"Is 'pets' a child of 'carnivores'? {carnivores.check_if_child(pets)}")
print(f"Is 'pets' a child of 'domestic'? {domestic.check_if_child(pets)}")
print(f"Is 'pets' a child of 'aquatic'? {aquatic.check_if_child(pets)}")

print(f"\nIs 'aquatic_carnivores' a child of 'carnivores'? {carnivores.check_if_child(aquatic_carnivores)}")
print(f"Is 'aquatic_carnivores' a child of 'aquatic'? {aquatic.check_if_child(aquatic_carnivores)}")
print(f"Is 'aquatic_carnivores' a child of 'domestic'? {domestic.check_if_child(aquatic_carnivores)}")

# Get group lengths with recursion (including child groups)
group_lengths_recursive = multi_parent_list.get_group_lengths(recurse=True)
print(f"\nGroup lengths (recursive): {group_lengths_recursive}")
````

### Requiring Groups

You can ensure that a group exists, creating it if necessary:

````python
# Create a new GroupedList
required_list = GroupedList()

# Require a group (creates it if it doesn't exist)
colors = required_list.require_group("colors")
colors.extend(["Red", "Green", "Blue"])
print(f"Colors group: {colors}")

# Require a group that already exists (returns the existing group)
existing_colors = required_list.require_group("colors")
print(f"Existing colors group: {existing_colors}")
print(f"Are they the same object? {colors is existing_colors}")

# Require a nested group using dot notation
nested_colors = required_list.require_group("colors_primary")
nested_colors.extend(["Red", "Yellow", "Blue"])
print(f"\nNested colors group: {nested_colors}")

# Check the structure
print(f"Main list: {required_list}")
print(f"Colors group: {colors}")
print(f"Primary colors group: {required_list.get_group('colors_primary')}")
````

### Flattening the List

You can flatten the grouped list into a simple list or tuple:

````python
# Create a GroupedList with multiple groups
flat_list = GroupedList(["Item 1", "Item 2", "Item 3"])
fruits = flat_list.create_group("fruits")
fruits.extend(["Apple", "Banana", "Orange"])
vegetables = flat_list.create_group("vegetables")
vegetables.extend(["Carrot", "Broccoli", "Spinach"])

# Flatten the list
flat_tuple = flat_list.as_flat_tuple()
print(f"Flat tuple: {flat_tuple}")

flat_list_result = flat_list.as_flat_list()
print(f"Flat list: {flat_list_result}")
````

### Task Management System

`GroupedList` is perfect for implementing a simple task management system:

````python
class TaskManager:
    """A simple task management system using GroupedList."""

    def __init__(self) -> None:
        self.tasks = GroupedList()

        # Create default categories
        self.tasks.create_group("work")
        self.tasks.create_group("personal")
        self.tasks.create_group("shopping")

        # Create priority groups
        self.tasks.create_group("high_priority")
        self.tasks.create_group("medium_priority")
        self.tasks.create_group("low_priority")

    def add_task(self, task, category=None, priority=None) -> None:
        """Add a task to the specified category and priority."""
        # Add to the main list
        self.tasks.append(task)

        # Add to the category if specified
        if category:
            category_group = self.tasks.require_group(category)
            category_group.append(task)

        # Add to the priority if specified
        if priority:
            priority_group = self.tasks.require_group(f"{priority}_priority")
            priority_group.append(task)

    def get_tasks_by_category(self, category):
        """Get all tasks in a specific category."""
        try:
            return self.tasks.get_group(category)
        except KeyError:
            return []

    def get_tasks_by_priority(self, priority):
        """Get all tasks with a specific priority."""
        try:
            return self.tasks.get_group(f"{priority}_priority")
        except KeyError:
            return []

    def mark_completed(self, task) -> None:
        """Mark a task as completed by moving it to the completed group."""
        completed_group = self.tasks.require_group("completed")

        # Find the task in the main list
        try:
            self.tasks.index(task)
            # Add to completed group
            completed_group.append(task)
            # No need to remove from main list as it's still part of the overall tasks
        except ValueError:
            print(f"Task '{task}' not found")

    def get_all_tasks(self):
        """Get all tasks."""
        return self.tasks

    def get_completed_tasks(self):
        """Get all completed tasks."""
        try:
            return self.tasks.get_group("completed")
        except KeyError:
            return []

    def __str__(self) -> str:
        result = "Task Manager:\n"
        result += f"  All tasks: {list(self.tasks)}\n"

        for group_name in [
            "work",
            "personal",
            "shopping",
            "high_priority",
            "medium_priority",
            "low_priority",
            "completed",
        ]:
            try:
                group = self.tasks.get_group(group_name)
                result += f"  {group_name}: {list(group)}\n"
            except KeyError:
                result += f"  {group_name}: []\n"

        return result


# Create a task manager
task_manager = TaskManager()

# Add some tasks
task_manager.add_task("Finish project report", category="work", priority="high")
task_manager.add_task("Buy groceries", category="shopping", priority="medium")
task_manager.add_task("Call mom", category="personal", priority="medium")
task_manager.add_task("Pay bills", category="personal", priority="high")
task_manager.add_task("Buy birthday gift", category="shopping", priority="low")

print(task_manager)

# Get tasks by category
work_tasks = task_manager.get_tasks_by_category("work")
print(f"\nWork tasks: {work_tasks}")

# Get tasks by priority
high_priority_tasks = task_manager.get_tasks_by_priority("high")
print(f"High priority tasks: {high_priority_tasks}")

# Mark a task as completed
task_manager.mark_completed("Buy groceries")
print(f"\nCompleted tasks: {task_manager.get_completed_tasks()}")

# Final state
print(f"\n{task_manager}")
````

### Document Organization System

Another practical use case for `GroupedList` is implementing a document organization system:

````python
class DocumentOrganizer:
    """A document organization system using GroupedList."""

    def __init__(self) -> None:
        self.documents = GroupedList()

        # Create default categories
        self.documents.create_group("reports")
        self.documents.create_group("presentations")
        self.documents.create_group("spreadsheets")

        # Create status groups
        self.documents.create_group("draft")
        self.documents.create_group("review")
        self.documents.create_group("final")

        # Create access level groups
        self.documents.create_group("public")
        self.documents.create_group("internal")
        self.documents.create_group("confidential")

    def add_document(self, document, category=None, status=None, access_level=None) -> None:
        """Add a document with the specified category, status, and access level."""
        # Add to the main list
        self.documents.append(document)

        # Add to the category if specified
        if category:
            category_group = self.documents.require_group(category)
            category_group.append(document)

        # Add to the status if specified
        if status:
            status_group = self.documents.require_group(status)
            status_group.append(document)

        # Add to the access level if specified
        if access_level:
            access_group = self.documents.require_group(access_level)
            access_group.append(document)

    def create_subcategory(self, parent_category, subcategory_name):
        """Create a subcategory under a parent category."""
        parent = self.documents.require_group(parent_category)
        return parent.create_group(subcategory_name)

    def get_documents_by_category(self, category):
        """Get all documents in a specific category."""
        try:
            return self.documents.get_group(category)
        except KeyError:
            return []

    def get_documents_by_status(self, status):
        """Get all documents with a specific status."""
        try:
            return self.documents.get_group(status)
        except KeyError:
            return []

    def get_documents_by_access_level(self, access_level):
        """Get all documents with a specific access level."""
        try:
            return self.documents.get_group(access_level)
        except KeyError:
            return []

    def update_document_status(self, document, new_status) -> None:
        """Update the status of a document."""
        # Remove from all status groups
        for status in ["draft", "review", "final"]:
            try:
                status_group = self.documents.get_group(status)
                if document in status_group:
                    status_group.remove(document)
            except (KeyError, ValueError):
                pass

        # Add to the new status group
        status_group = self.documents.require_group(new_status)
        status_group.append(document)

    def __str__(self) -> str:
        result = "Document Organizer:\n"
        result += f"  All documents: {list(self.documents)}\n"

        for group_name in [
            "reports",
            "presentations",
            "spreadsheets",
            "draft",
            "review",
            "final",
            "public",
            "internal",
            "confidential",
        ]:
            try:
                group = self.documents.get_group(group_name)
                result += f"  {group_name}: {list(group)}\n"
            except KeyError:
                result += f"  {group_name}: []\n"

        return result


# Create a document organizer
doc_organizer = DocumentOrganizer()

# Add some documents
doc_organizer.add_document("Q1 Financial Report", category="reports", status="draft", access_level="confidential")
doc_organizer.add_document("Company Overview", category="presentations", status="final", access_level="public")
doc_organizer.add_document("Budget Forecast", category="spreadsheets", status="review", access_level="internal")
doc_organizer.add_document("Marketing Plan", category="reports", status="draft", access_level="internal")
doc_organizer.add_document("Product Roadmap", category="presentations", status="review", access_level="internal")

print(doc_organizer)

# Create subcategories
financial_reports = doc_organizer.create_subcategory("reports", "financial_reports")
financial_reports.append("Q1 Financial Report")  # Add existing document to subcategory
financial_reports.append("Annual Report")  # Add new document to subcategory

# Get documents by category
reports = doc_organizer.get_documents_by_category("reports")
print(f"\nReports: {reports}")
financial_reports_result = doc_organizer.get_documents_by_category("financial_reports")
print(f"Financial Reports: {financial_reports_result}")

# Get documents by status
draft_documents = doc_organizer.get_documents_by_status("draft")
print(f"\nDraft documents: {draft_documents}")

# Update document status
doc_organizer.update_document_status("Q1 Financial Report", "final")
print("\nAfter updating Q1 Financial Report status to final:")
print(f"Draft documents: {doc_organizer.get_documents_by_status('draft')}")
print(f"Final documents: {doc_organizer.get_documents_by_status('final')}")
````

## API Highlights

Here are the key components of the `GroupedList` module API:

### GroupedList
- `__init__(items=None, parent=None, parents=None, init=True)`: Initialize a new GroupedList instance
- `create_group(name, items=None)`: Create a new group with the specified name
- `require_group(name, items=None)`: Get a group with the specified name, creating it if it doesn't exist
- `remove_group(group)`: Remove a group from the list
- `get_group(name)`: Get a group by name
- `add_group(group, name)`: Add an existing group to the list
- `get_group_lengths(recurse=False)`: Get the lengths of all groups
- `get_length()`: Get the total length of the list, including all groups
- `check_if_child(other)`: Check if another group is a child of this group
- `check_if_parent(other)`: Check if another group is a parent of this group
- `as_flat_tuple()`: Get a tuple containing all items in the list and its groups
- `as_flat_list()`: Get a list containing all items in the list and its groups
- All standard list methods (`append`, `insert`, `pop`, `remove`, etc.) with optional `group` parameter

For more detailed information, consult the full API documentation.

## Troubleshooting / FAQs

### Q: When should I use GroupedList instead of a regular list or dictionary?

A: `GroupedList` is useful when you need to:
- Organize items into logical groups while maintaining a flat list structure
- Perform operations on specific groups of items
- Create hierarchical relationships between groups
- Maintain multiple categorizations of the same items

If you only need simple categorization without the need for a unified list, a dictionary of lists might be simpler. If you don't need categorization at all, a regular list is sufficient.

### Q: How does GroupedList handle copying and deep copying?

A: `GroupedList` inherits from `BaseList`, which properly implements Python's copy protocol through `__copy__` and `__deepcopy__` methods. This ensures that both the list and its groups are properly copied.

### Q: Can an item belong to multiple groups?

A: Yes, the same item can be added to multiple groups. This is one of the key features of `GroupedList`, allowing for multiple categorizations of the same items.

### Q: How do I access items in a nested group?

A: You can access items in a nested group using dot notation with the `get_group` method:

```python
nested_group = grouped_list.get_group("parent.child")
```

Alternatively, you can access the parent group first, then the child group:

```python
parent_group = grouped_list.get_group("parent")
child_group = parent_group.get_group("child")
```
