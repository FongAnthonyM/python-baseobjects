# Using the CircularDoublyLinkedContainer Module in baseobjects

## Introduction

The `CircularDoublyLinkedContainer` module provides a circular doubly linked list implementation in the baseobjects package. It consists of two main classes: `LinkedNode` and `CircularDoublyLinkedContainer`. This data structure allows for efficient insertions and deletions at any position, as well as bidirectional traversal.

A circular doubly linked list is a linked data structure where each node contains a reference to both the next and previous nodes, and the last node points back to the first node, forming a circle. This circular nature allows for continuous traversal in both directions without reaching an end.

- Traversing the list in both directions
- Advanced features and use cases

## Importing the Module

```
from baseobjects.collections.circulardoublylinkedcontainer import CircularDoublyLinkedContainer, LinkedNode
import copy
```

## Core Functionality

The `CircularDoublyLinkedContainer` class provides a circular doubly linked list implementation. It consists of `LinkedNode` objects that are connected in a circular fashion, with each node pointing to both its next and previous nodes.

### Key Features

1. **Circular Structure**: The last node points back to the first node, and the first node points back to the last node
2. **Bidirectional Traversal**: You can traverse the list in both forward and backward directions
3. **Efficient Insertions and Deletions**: Operations at any position are efficient
4. **Shift Operations**: The list can be shifted left or right, changing the starting point

Let's create a simple `CircularDoublyLinkedContainer` and explore its functionality:

```
# Create an empty container
container = CircularDoublyLinkedContainer()
print(f"Empty container: {container}")
print(f"Is empty: {container.is_empty}")

# Add some elements
container.append("A")
container.append("B")
container.append("C")
container.append("D")
print(f"Container after adding elements: {container}")
print(f"Length: {len(container)}")
print(f"Is empty: {container.is_empty}")
```

### Accessing Elements

You can access elements in the container using indexing, similar to a regular list:

```
# Access elements by index
print(f"First element: {container[0]}")
print(f"Second element: {container[1]}")
print(f"Last element: {container[-1]}")

# Due to the circular nature, you can use indices beyond the length
print(f"container[4] (wraps around): {container[4]}")
print(f"container[5] (wraps around): {container[5]}")
print(f"container[-5] (wraps around): {container[-5]}")
```

### Iterating Through the Container

You can iterate through the container like a regular list:

```
# Iterate through the container
print("Iterating through the container:")
for item in container:
    print(item)
```

### Modifying the Container

You can modify the container using various methods:

```
# Insert an element at a specific position
container.insert("X", 2)
print(f"After inserting 'X' at position 2: {container}")

# Remove an element
popped_item = container.pop(1)
print(f"Popped item: {popped_item}")
print(f"After popping element at position 1: {container}")

# Clear the container
container_copy = copy.copy(container)
container_copy.clear()
print(f"After clearing (copy): {container_copy}")
print(f"Original container: {container}")
```

## Module Interaction

The `CircularDoublyLinkedContainer` module interacts with other modules in the baseobjects package, particularly `BaseObject` and `BaseReducible`. These interactions provide enhanced functionality for the circular doubly linked list.

### Interaction with BaseObject

`CircularDoublyLinkedContainer` inherits from `BaseObject`, which means it also inherits all the functionality of `BaseObject`, including copying and deep copying:

```
# Create a copy of the container
container_copy = container.copy()
print(f"Original: {container}")
print(f"Copy: {container_copy}")
print(f"Are they the same object? {container is container_copy}")

# Modify the copy
container_copy.append("Y")
print(f"\nAfter modifying the copy:")
print(f"Original: {container}")
print(f"Copy: {container_copy}")
```

### Deep Copying

`CircularDoublyLinkedContainer` also supports deep copying through inheritance from `BaseObject`:

```
# Create a container with complex objects
complex_container = CircularDoublyLinkedContainer()
complex_container.append([1, 2, 3])
complex_container.append({"key": "value"})
print(f"Complex container: {complex_container}")

# Create a deep copy
complex_container_deepcopy = copy.deepcopy(complex_container)
print(f"Deep copy: {complex_container_deepcopy}")
print(f"Are they the same object? {complex_container is complex_container_deepcopy}")

# Modify the nested object in the deep copy
complex_container_deepcopy[0].data.append(4)
complex_container_deepcopy[1].data["new_key"] = "new_value"

print(f"\nAfter modifying the deep copy:")
print(f"Original first element: {complex_container[0]}")
print(f"Original second element: {complex_container[1]}")
print(f"Deep copy first element: {complex_container_deepcopy[0]}")
print(f"Deep copy second element: {complex_container_deepcopy[1]}")
```

## Advanced Features

The `CircularDoublyLinkedContainer` class provides several advanced features that make it powerful for specific use cases.

### Shifting the Container

You can shift the container left or right, changing the starting point of the list:

```
# Create a new container
shift_container = CircularDoublyLinkedContainer()
shift_container.append("A")
shift_container.append("B")
shift_container.append("C")
shift_container.append("D")
shift_container.append("E")
print(f"Original container: {shift_container}")

# Shift left
shift_container.shift_left(2)
print(f"After shifting left by 2: {shift_container}")

# Shift right
shift_container.shift_right(3)
print(f"After shifting right by 3: {shift_container}")

# Using the shift operators
shifted_left = shift_container << 1
print(f"After using << operator: {shifted_left}")

shifted_right = shift_container >> 1
print(f"After using >> operator: {shifted_right}")
```

### Bidirectional Iteration

You can iterate through the container in both forward and backward directions:

```
# Create a new container
iter_container = CircularDoublyLinkedContainer()
iter_container.append("A")
iter_container.append("B")
iter_container.append("C")
iter_container.append("D")
print(f"Container: {iter_container}")

# Forward iteration
print("\nForward iteration:")
for item in iter_container.forward_iter():
    print(item)

# Reverse iteration
print("\nReverse iteration:")
for item in iter_container.reverse_iter():
    print(item)
```

### Infinite Cycling

You can create infinite iterators that cycle through the container:

```
# Create a new container
cycle_container = CircularDoublyLinkedContainer()
cycle_container.append("A")
cycle_container.append("B")
cycle_container.append("C")
print(f"Container: {cycle_container}")

# Forward cycle
print("\nForward cycle (first 10 items):")
forward_cycle = cycle_container.forward_cycle()
for _ in range(10):
    print(next(forward_cycle))

# Reverse cycle
print("\nReverse cycle (first 10 items):")
reverse_cycle = cycle_container.reverse_cycle()
for _ in range(10):
    print(next(reverse_cycle))
```

### Working with Nodes Directly

You can work with the `LinkedNode` objects directly for more advanced operations:

```
# Create a new container
node_container = CircularDoublyLinkedContainer()
node_container.append("A")
node_container.append("B")
node_container.append("C")
print(f"Container: {node_container}")

# Create a new node
new_node = LinkedNode("D")
print(f"New node data: {new_node.data}")

# Add the node to the container
node_container.append(new_node)
print(f"Container after adding node: {node_container}")

# Get the last node
last_node = node_container.last_node
print(f"Last node data: {last_node.data}")

# Move a node to the start
node_container.move_node_start(last_node)
print(f"Container after moving last node to start: {node_container}")

# Remove a node
node_to_remove = node_container.get_item(1)
node_container.remove_node(node_to_remove)
print(f"Container after removing node at index 1: {node_container}")
```

## Examples

### Implementing a Circular Buffer

A circular buffer is a data structure that uses a fixed-size buffer as if it were connected end-to-end. This is a perfect use case for `CircularDoublyLinkedContainer`:

```
class CircularBuffer:
    """A fixed-size buffer that overwrites old data when full."""
    
    def __init__(self, capacity):
        self.container = CircularDoublyLinkedContainer()
        self.capacity = capacity
        self.size = 0
    
    def write(self, item):
        """Write an item to the buffer."""
        if self.size < self.capacity:
            self.container.append(item)
            self.size += 1
        else:
            # Overwrite the oldest item
            self.container[0].data = item
            # Shift the buffer so the next oldest item becomes the first
            self.container.shift_left(1)
    
    def read(self):
        """Read all items from the buffer."""
        return list(self.container)
    
    def __str__(self):
        return str(self.read())

# Create a circular buffer with capacity 3
buffer = CircularBuffer(3)

# Write some items
buffer.write("A")
print(f"Buffer after writing 'A': {buffer}")

buffer.write("B")
print(f"Buffer after writing 'B': {buffer}")

buffer.write("C")
print(f"Buffer after writing 'C': {buffer}")

# Buffer is now full, writing more items will overwrite old ones
buffer.write("D")
print(f"Buffer after writing 'D': {buffer}")

buffer.write("E")
print(f"Buffer after writing 'E': {buffer}")

buffer.write("F")
print(f"Buffer after writing 'F': {buffer}")
```

### Implementing a Round-Robin Scheduler

A round-robin scheduler is a simple scheduling algorithm that assigns time slices to each process in equal portions and in circular order. This is another good use case for `CircularDoublyLinkedContainer`:

```
class RoundRobinScheduler:
    """A simple round-robin scheduler."""
    
    def __init__(self):
        self.processes = CircularDoublyLinkedContainer()
        self.current_index = 0
    
    def add_process(self, process):
        """Add a process to the scheduler."""
        self.processes.append(process)
    
    def remove_process(self, process):
        """Remove a process from the scheduler."""
        for i, p in enumerate(self.processes):
            if p == process:
                self.processes.pop(i)
                break
    
    def next_process(self):
        """Get the next process in the round-robin schedule."""
        if not self.processes:
            return None
        
        process = self.processes[self.current_index]
        self.processes.shift_left(1)
        return process
    
    def __str__(self):
        return f"RoundRobinScheduler(processes={list(self.processes)})"

# Create a round-robin scheduler
scheduler = RoundRobinScheduler()

# Add some processes
scheduler.add_process("Process A")
scheduler.add_process("Process B")
scheduler.add_process("Process C")
print(f"Scheduler: {scheduler}")

# Run the scheduler for a few cycles
print("\nRunning the scheduler:")
for _ in range(10):
    process = scheduler.next_process()
    print(f"Running {process}")

# Remove a process
scheduler.remove_process("Process B")
print(f"\nAfter removing 'Process B': {scheduler}")

# Run the scheduler again
print("\nRunning the scheduler again:")
for _ in range(6):
    process = scheduler.next_process()
    print(f"Running {process}")
```

## API Highlights

Here are the key components of the `CircularDoublyLinkedContainer` module API:

### LinkedNode
- `__init__(data=None, previous=None, next_=None, *args, init=True, **kwargs)`: Initialize a new LinkedNode instance
- `previous`: Property to get or set the previous node
- `next`: Property to get or set the next node
- `data`: The data stored in the node

- `last_node`: Property that gets the last node in the container

- `forward_iter()`: Iterate through the container in forward direction
- `reverse_iter()`: Iterate through the container in reverse direction
- `forward_cycle()`: Create an infinite iterator that cycles through the container in forward direction
- `reverse_cycle()`: Create an infinite iterator that cycles through the container in reverse direction
- `move_node_start(node)`: Move a node to the start of the container
- `move_node_end(node)`: Move a node to the end of the container

## Troubleshooting / FAQs

### Q: When should I use CircularDoublyLinkedContainer instead of a regular list?

A: `CircularDoublyLinkedContainer` is useful when you need:
- Efficient insertions and deletions at any position
- Bidirectional traversal
- Circular behavior (wrapping around from the end to the beginning)
- Shifting the starting point of the list

Regular lists are more efficient for random access and when you don't need the circular behavior.

### Q: How does CircularDoublyLinkedContainer handle copying and deep copying?

A: `CircularDoublyLinkedContainer` inherits from `BaseObject`, which properly implements Python's copy protocol through `__copy__` and `__deepcopy__` methods. This ensures that both the container and its nodes are properly copied, maintaining the circular structure.

### Q: Can I use negative indices with CircularDoublyLinkedContainer?

A: Yes, negative indices work similar to regular lists, with -1 referring to the last item, -2 to the second-to-last item, and so on. Additionally, due to the circular nature, you can use indices beyond the length of the container, and they will wrap around.
