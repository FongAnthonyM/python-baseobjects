# Using the TimedDict Module in baseobjects

This type of dictionary is particularly useful for implementing caches with automatic expiration, session storage with timeouts, or any scenario where data should be automatically cleared after a certain period of inactivity.

- Controlling the timing behavior
- Advanced features and use cases

## Importing the Module

```
from baseobjects.collections.timeddict import TimedDict
import time
import copy
```

## Core Functionality

The `TimedDict` class provides a dictionary-like container that automatically clears its contents after a specified time has elapsed. Let's explore its core functionality.

### Key Features

1. **Dictionary-Like Behavior**: Provides all standard dictionary operations
2. **Automatic Expiration**: Clears contents after a specified time period
3. **Configurable Lifetime**: Set the time period after which the dictionary is cleared
4. **Timer Control**: Pause and resume the timer as needed
5. **Inheritance from BaseDict**: Provides proper copying and deep copying support

Let's create a simple `TimedDict` and explore its functionality:

```
# Create a TimedDict with a 2-second lifetime
timed_dict = TimedDict()
timed_dict.lifetime = 2  # Set lifetime to 2 seconds
print(f"Empty TimedDict: {timed_dict}")
print(f"Lifetime: {timed_dict.lifetime} seconds")
print(f"Is timed: {timed_dict.is_timed}")

# Add some items
timed_dict["a"] = 1
timed_dict["b"] = 2
timed_dict["c"] = 3
print(f"TimedDict with items: {timed_dict}")

# Access items
print(f"Value of 'a': {timed_dict['a']}")
print(f"Value of 'b': {timed_dict['b']}")

# Wait for the dictionary to expire
print("\nWaiting for 2.5 seconds...")
time.sleep(2.5)

# Try to access the dictionary after expiration
print(f"TimedDict after expiration: {timed_dict}")
```

### Dictionary Methods

`TimedDict` inherits all the standard dictionary methods from `BaseDict`:

```
# Create a new TimedDict with a longer lifetime
timed_dict = TimedDict()
timed_dict.lifetime = 10  # Set lifetime to 10 seconds

# Add some items
timed_dict["a"] = 1
timed_dict["b"] = 2
timed_dict["c"] = 3
print(f"TimedDict with items: {timed_dict}")

# Get all keys
print(f"timed_dict.keys(): {timed_dict.keys()}")

# Get all values
print(f"timed_dict.values(): {timed_dict.values()}")

# Get all items
print(f"timed_dict.items(): {timed_dict.items()}")

# Get a value with a default if the key doesn't exist
print(f"timed_dict.get('a'): {timed_dict.get('a')}")
print(f"timed_dict.get('z', 'default'): {timed_dict.get('z', 'default')}")

# Check if a key exists
print(f"'b' in timed_dict: {'b' in timed_dict}")
print(f"'z' in timed_dict: {'z' in timed_dict}")

# Update the dictionary
timed_dict.update({"d": 4, "e": 5})
print(f"After update: {timed_dict}")

# Remove an item
del timed_dict["e"]
print(f"After deleting 'e': {timed_dict}")

# Pop an item
popped_value = timed_dict.pop("d")
print(f"Popped value: {popped_value}")
print(f"After pop: {timed_dict}")
```

### Expiration Behavior

Let's explore the expiration behavior of `TimedDict` in more detail:

```
# Create a TimedDict with a 3-second lifetime
expiration_dict = TimedDict()
expiration_dict.lifetime = 3
print(f"Initial expiration_dict: {expiration_dict}")

# Add some items
expiration_dict["a"] = 1
expiration_dict["b"] = 2
print(f"After adding items: {expiration_dict}")

# Wait for 1.5 seconds (half the lifetime)
print("\nWaiting for 1.5 seconds...")
time.sleep(1.5)

# Access the dictionary (this should reset the expiration timer)
print(f"After 1.5 seconds: {expiration_dict}")

# Add another item (this should also reset the expiration timer)
expiration_dict["c"] = 3
print(f"After adding 'c': {expiration_dict}")

# Wait for 2 seconds (less than the lifetime)
print("\nWaiting for 2 seconds...")
time.sleep(2)

# The dictionary should still be valid
print(f"After 2 more seconds: {expiration_dict}")

# Wait for 1.5 more seconds (exceeding the lifetime)
print("\nWaiting for 1.5 more seconds...")
time.sleep(1.5)

# The dictionary should be cleared
print(f"After exceeding lifetime: {expiration_dict}")
```

### Manually Clearing the Dictionary

You can manually clear the `TimedDict` at any time:

```
# Create a TimedDict with a long lifetime
manual_dict = TimedDict()
manual_dict.lifetime = 60  # Set lifetime to 60 seconds

# Add some items
manual_dict["a"] = 1
manual_dict["b"] = 2
manual_dict["c"] = 3
print(f"Initial manual_dict: {manual_dict}")

# Manually clear the dictionary
manual_dict.clear()
print(f"After clear(): {manual_dict}")

# Add new items
manual_dict["x"] = 10
manual_dict["y"] = 20
print(f"After adding new items: {manual_dict}")
```

## Module Interaction

The `TimedDict` module interacts with other modules in the baseobjects package, particularly `BaseDict`. These interactions provide enhanced functionality for the timed dictionary.

### Interaction with BaseDict

`TimedDict` inherits from `BaseDict`, which means it also inherits all the functionality of `BaseDict`, including copying and deep copying:

```
# Create a TimedDict
timed_dict = TimedDict()
timed_dict.lifetime = 30  # Set lifetime to 30 seconds
timed_dict["a"] = 1
timed_dict["b"] = 2
timed_dict["c"] = 3

# Create a copy of the TimedDict
timed_dict_copy = timed_dict.copy()
print(f"Original: {timed_dict}")
print(f"Copy: {timed_dict_copy}")
print(f"Are they the same object? {timed_dict is timed_dict_copy}")
print(f"Original lifetime: {timed_dict.lifetime}")
print(f"Copy lifetime: {timed_dict_copy.lifetime}")

# Modify the copy
timed_dict_copy["d"] = 4
timed_dict_copy.lifetime = 10
print("\nAfter modifying the copy:")
print(f"Original: {timed_dict}")
print(f"Copy: {timed_dict_copy}")
print(f"Original lifetime: {timed_dict.lifetime}")
print(f"Copy lifetime: {timed_dict_copy.lifetime}")
```

### Deep Copying

`TimedDict` also supports deep copying through inheritance from `BaseDict`:

```
# Create a TimedDict with nested structures
nested_dict = TimedDict()
nested_dict.lifetime = 30  # Set lifetime to 30 seconds
nested_dict["simple"] = 1
nested_dict["list"] = [1, 2, 3]
nested_dict["dict"] = {"a": 1, "b": 2}
print(f"Nested dict: {nested_dict}")

# Create a deep copy
nested_dict_deepcopy = copy.deepcopy(nested_dict)
print(f"Deep copy: {nested_dict_deepcopy}")
print(f"Are they the same object? {nested_dict is nested_dict_deepcopy}")
print(f"Is the nested list the same object? {nested_dict['list'] is nested_dict_deepcopy['list']}")
print(f"Is the nested dict the same object? {nested_dict['dict'] is nested_dict_deepcopy['dict']}")

# Modify the nested structures in the deep copy
nested_dict_deepcopy["list"].append(4)
nested_dict_deepcopy["dict"]["c"] = 3

print("\nAfter modifying the deep copy:")
print(f"Original list: {nested_dict['list']}")
print(f"Original dict: {nested_dict['dict']}")
print(f"Deep copy list: {nested_dict_deepcopy['list']}")
print(f"Deep copy dict: {nested_dict_deepcopy['dict']}")
```

## Advanced Features

The `TimedDict` class provides several advanced features that make it powerful for specific use cases.

### Pausing the Timer

You can pause the timer to prevent the dictionary from expiring:

```
# Create a TimedDict with a 3-second lifetime
pause_dict = TimedDict()
pause_dict.lifetime = 3
pause_dict["a"] = 1
pause_dict["b"] = 2
print(f"Initial pause_dict: {pause_dict}")

# Pause the timer
print("\nPausing the timer...")
pause_dict.is_timed = False

# Wait for 4 seconds (longer than the lifetime)
print("Waiting for 4 seconds...")
time.sleep(4)

# The dictionary should still be valid because the timer is paused
print(f"After 4 seconds with timer paused: {pause_dict}")

# Resume the timer
print("\nResuming the timer...")
pause_dict.is_timed = True

# Wait for 1.5 seconds (half the lifetime)
print("Waiting for 1.5 seconds...")
time.sleep(1.5)

# The dictionary should still be valid
print(f"After 1.5 seconds with timer resumed: {pause_dict}")

# Wait for 2 more seconds (exceeding the lifetime)
print("\nWaiting for 2 more seconds...")
time.sleep(2)

# The dictionary should be cleared
print(f"After exceeding lifetime: {pause_dict}")
```

### Using Context Managers

`TimedDict` provides context managers for pausing the timer:

```
# Create a TimedDict with a 2-second lifetime
context_dict = TimedDict()
context_dict.lifetime = 2
context_dict["a"] = 1
context_dict["b"] = 2
print(f"Initial context_dict: {context_dict}")

# Use the pause_timer context manager
print("\nEntering pause_timer context...")
with context_dict.pause_timer():
    print(f"Inside context: {context_dict}")

    # Wait for 3 seconds (longer than the lifetime)
    print("Waiting for 3 seconds...")
    time.sleep(3)

    print(f"After 3 seconds inside context: {context_dict}")

print("Exiting pause_timer context...")

# Wait for 1 second
print("\nWaiting for 1 second after context...")
time.sleep(1)

# The dictionary should still be valid because the timer was paused during the wait
print(f"After 1 second outside context: {context_dict}")

# Wait for 1.5 more seconds (exceeding the lifetime)
print("\nWaiting for 1.5 more seconds...")
time.sleep(1.5)

# The dictionary should be cleared
print(f"After exceeding lifetime: {context_dict}")
```

### Using pause_reset_timer Context Manager

The `pause_reset_timer` context manager pauses the timer and resets the expiration when exiting the context:

```
# Create a TimedDict with a 2-second lifetime
reset_dict = TimedDict()
reset_dict.lifetime = 2
reset_dict["a"] = 1
reset_dict["b"] = 2
print(f"Initial reset_dict: {reset_dict}")

# Use the pause_reset_timer context manager
print("\nEntering pause_reset_timer context...")
with reset_dict.pause_reset_timer():
    print(f"Inside context: {reset_dict}")

    # Wait for 3 seconds (longer than the lifetime)
    print("Waiting for 3 seconds...")
    time.sleep(3)

    print(f"After 3 seconds inside context: {reset_dict}")

print("Exiting pause_reset_timer context...")

# The expiration should be reset when exiting the context
print("\nWaiting for 1.5 seconds after context...")
time.sleep(1.5)

# The dictionary should still be valid because the expiration was reset
print(f"After 1.5 seconds outside context: {reset_dict}")

# Wait for 1 more second (exceeding the lifetime)
print("\nWaiting for 1 more second...")
time.sleep(1)

# The dictionary should be cleared
print(f"After exceeding lifetime: {reset_dict}")
```

### Customizing Clear Condition

You can subclass `TimedDict` to customize the condition for clearing the dictionary:

```
class CustomTimedDict(TimedDict):
    """A TimedDict that clears based on a custom condition."""

    def __init__(self, max_items=5, **kwargs) -> None:
        super().__init__(**kwargs)
        self.max_items = max_items

    def clear_condition(self, *args, **kwargs):
        """Clear the dictionary if it has more than max_items or if the timer expires."""
        # Check the standard time-based condition
        time_condition = super().clear_condition(*args, **kwargs)

        # Check if the dictionary has more than max_items
        size_condition = len(self._data) > self.max_items

        return time_condition or size_condition


# Create a CustomTimedDict with a 10-second lifetime and max_items=3
custom_dict = CustomTimedDict(max_items=3)
custom_dict.lifetime = 10
print(f"Initial custom_dict: {custom_dict}")

# Add some items
custom_dict["a"] = 1
custom_dict["b"] = 2
custom_dict["c"] = 3
print(f"After adding 3 items: {custom_dict}")

# Add a 4th item (exceeding max_items)
custom_dict["d"] = 4
print(f"After adding a 4th item: {custom_dict}")

# The dictionary should be cleared because it exceeded max_items
print(f"Dictionary after exceeding max_items: {custom_dict}")
```

## Examples

Let's explore some practical examples of using the `TimedDict` module.

### Implementing a Simple Cache

`TimedDict` is perfect for implementing a simple cache with automatic expiration:

```
class SimpleCache:
    """A simple cache with automatic expiration."""

    def __init__(self, lifetime=60) -> None:
        self.cache = TimedDict()
        self.cache.lifetime = lifetime
        self.hits = 0
        self.misses = 0

    def get(self, key, default=None):
        """Get a value from the cache."""
        if key in self.cache:
            self.hits += 1
            return self.cache[key]
        else:
            self.misses += 1
            return default

    def set(self, key, value) -> None:
        """Set a value in the cache."""
        self.cache[key] = value

    def clear(self) -> None:
        """Clear the cache."""
        self.cache.clear()

    def get_stats(self):
        """Get cache statistics."""
        total = self.hits + self.misses
        hit_rate = self.hits / total if total > 0 else 0
        return {
            "hits": self.hits,
            "misses": self.misses,
            "total": total,
            "hit_rate": hit_rate,
            "size": len(self.cache),
        }

    def __str__(self) -> str:
        stats = self.get_stats()
        return (f"SimpleCache(size={stats['size']}, "
                f"hits={stats['hits']}, misses={stats['misses']}, "
                f"hit_rate={stats['hit_rate']:.2f})")


# Create a simple cache with a 5-second lifetime
cache = SimpleCache(lifetime=5)


# Simulate a function that's expensive to compute
def expensive_computation(x):
    print(f"Computing result for {x}...")
    time.sleep(1)  # Simulate a time-consuming operation
    return x * x


# Function that uses the cache
def cached_computation(x):
    result = cache.get(x)
    if result is None:
        result = expensive_computation(x)
        cache.set(x, result)
    return result


# Use the cached function
print("\nFirst round of computations:")
for x in [1, 2, 3, 1, 2, 3]:
    result = cached_computation(x)
    print(f"Result for {x}: {result}")
    print(f"Cache: {cache}")

# Wait for 3 seconds (less than the lifetime)
print("\nWaiting for 3 seconds...")
time.sleep(3)

# Use the cached function again
print("\nSecond round of computations:")
for x in [1, 2, 3, 4, 5]:
    result = cached_computation(x)
    print(f"Result for {x}: {result}")
    print(f"Cache: {cache}")

# Wait for 3 more seconds (exceeding the lifetime)
print("\nWaiting for 3 more seconds...")
time.sleep(3)

# Use the cached function again
print("\nThird round of computations:")
for x in [1, 2, 3]:
    result = cached_computation(x)
    print(f"Result for {x}: {result}")
    print(f"Cache: {cache}")
```

### Implementing a Session Store

Another practical use case for `TimedDict` is implementing a session store with automatic expiration:

```
class SessionStore:
    """A simple session store with automatic expiration."""

    def __init__(self, session_lifetime=1800) -> None:  # Default: 30 minutes
        self.sessions = TimedDict()
        self.sessions.lifetime = session_lifetime

    def create_session(self, user_id, data=None):
        """Create a new session for a user."""
        session_id = f"session_{user_id}_{int(time.time())}"
        session_data = {
            "user_id": user_id,
            "created_at": time.time(),
            "last_accessed": time.time(),
            "data": data or {},
        }
        self.sessions[session_id] = session_data
        return session_id

    def get_session(self, session_id):
        """Get a session by ID."""
        session = self.sessions.get(session_id)
        if session:
            # Update last accessed time
            session["last_accessed"] = time.time()
            self.sessions[session_id] = session
        return session

    def update_session(self, session_id, data) -> bool:
        """Update session data."""
        session = self.get_session(session_id)
        if session:
            session["data"].update(data)
            session["last_accessed"] = time.time()
            self.sessions[session_id] = session
            return True
        return False

    def delete_session(self, session_id) -> bool:
        """Delete a session."""
        if session_id in self.sessions:
            del self.sessions[session_id]
            return True
        return False

    def get_active_sessions(self):
        """Get all active sessions."""
        return self.sessions

    def __str__(self) -> str:
        return f"SessionStore(active_sessions={len(self.sessions)})"


# Create a session store with a 10-second session lifetime (for demonstration)
session_store = SessionStore(session_lifetime=10)
print(f"Initial session store: {session_store}")

# Create sessions for some users
session1 = session_store.create_session("user1", {"name": "Alice", "role": "admin"})
session2 = session_store.create_session("user2", {"name": "Bob", "role": "user"})
print(f"\nAfter creating sessions: {session_store}")

# Get session data
user1_session = session_store.get_session(session1)
print(f"\nUser 1 session: {user1_session}")

# Update session data
session_store.update_session(session1, {"last_page": "/dashboard"})
user1_session = session_store.get_session(session1)
print(f"\nUser 1 session after update: {user1_session}")

# Wait for 5 seconds (half the session lifetime)
print("\nWaiting for 5 seconds...")
time.sleep(5)

# Sessions should still be active
print(f"Session store after 5 seconds: {session_store}")
print(f"User 1 session still active: {session_store.get_session(session1) is not None}")
print(f"User 2 session still active: {session_store.get_session(session2) is not None}")

# Keep user1's session active by accessing it
session_store.get_session(session1)

# Wait for 6 more seconds (exceeding the lifetime for user2's session)
print("\nWaiting for 6 more seconds...")
time.sleep(6)

# User2's session should be expired, but user1's session should still be active
print(f"Session store after 11 seconds: {session_store}")
print(f"User 1 session still active: {session_store.get_session(session1) is not None}")
print(f"User 2 session still active: {session_store.get_session(session2) is not None}")

# Wait for 5 more seconds (exceeding the lifetime for user1's session)
print("\nWaiting for 5 more seconds...")
time.sleep(5)

# All sessions should be expired
print(f"Session store after 16 seconds: {session_store}")
```

## API Highlights

Here are the key components of the `TimedDict` module API:

### TimedDict
- `__init__(dict_=None, /, **kwargs)`: Initialize a new TimedDict instance
- `lifetime`: Property to get or set the period between dictionary resets in seconds
- `is_timed`: Boolean attribute that determines if the dictionary will be reset periodically
- `expiration`: Attribute that stores the next time the dictionary will be reset
- `clear()`: Clears the contents of the dictionary and resets the expiration
- `reset_expiration()`: Updates the expiration to a new future time
- `pause_timer()`: Context manager that pauses the timer until the context is exited
- `pause_reset_timer()`: Context manager that pauses the timer and resets the expiration when exiting
- `clear_condition(*args, **kwargs)`: Method that determines if the dictionary should be cleared
- `verify()`: Verifies if the dictionary should be cleared and then clears it if necessary
- All standard dictionary methods inherited from `BaseDict`

For more detailed information, consult the full API documentation.

## Troubleshooting / FAQs

### Q: When should I use TimedDict instead of a regular dictionary?

A: `TimedDict` is useful when you need a dictionary that automatically clears its contents after a specified time period. This is particularly useful for implementing caches with automatic expiration, session storage with timeouts, or any scenario where data should be automatically cleared after a certain period of inactivity.

### Q: How does TimedDict handle copying and deep copying?

A: `TimedDict` inherits from `BaseDict`, which properly implements Python's copy protocol through `__copy__` and `__deepcopy__` methods. This ensures that both the dictionary and its timing attributes are properly copied.

### Q: Does accessing the dictionary reset the expiration timer?

A: Yes, accessing the dictionary through the `data` property (which happens when you access items, iterate through the dictionary, etc.) calls the `verify()` method, which checks if the dictionary should be cleared. This effectively resets the expiration timer.

### Q: Can I disable the automatic expiration?

A: Yes, you can disable the automatic expiration by setting `is_timed` to `False` or by setting `lifetime` to `None`. You can also use the `pause_timer()` context manager to temporarily disable the expiration.
